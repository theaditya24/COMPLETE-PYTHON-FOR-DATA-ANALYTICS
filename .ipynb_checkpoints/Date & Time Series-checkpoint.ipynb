## Date & Time Data's

#### 
1. 'YYYY-MM-DD'  ---> Date Format
2. 'HH:MI:SS'    ---> Time Format
3. 'YYYY-MM-DD HH:MI:SS'  ---> TimeStamp
4. None ---> Python Empty Data
5. NaN  ---> Pandas Empty Data
6. NaT  ---> Date & Time Empty Data

```python
raise ZeroDivisionError('''time data "invalid_date" doesn't match format "%Y-%m-%d", at position 3. You might want to try:- passing `format` if your strings have a consistent format;
    - passing `format='ISO8601'` if your strings are all ISO8601 but not necessarily in exactly the same format;
    - passing `format='mixed'`, and the format will be inferred for each element individually. You might want to use `dayfirst` alongside this.''')
```
