## What is an API?

API stands for **`Application Programming Interface`**.

An API is a way for one software/application to communicate with another software/application and request or exchange data.

For Data Analytics, APIs are commonly used as external data sources.

### Why APIs Exist

APIs allow applications to communicate with external systems without requiring direct access to their internal databases.

Basic workflow: Client → API → Server → Data/Database → API → Client

APIs can control:
- What data can be accessed
- Who can access the data
- How requests should be made
- Authentication
- Rate limits
- Available functionality

### API in Data Analytics

A Data Analyst can use an API to collect data and then analyze it using Python.

Typical workflow: Python → API → JSON → Pandas → Data Cleaning → EDA → Visualization → Business Insights

### API vs Website

| Website | API |
|---|---|
| Mainly designed for humans | Mainly designed for applications |
| Usually displays information visually | Usually returns structured data |
| Browser is commonly used | Python/programs can be used |
| HTML/CSS/JavaScript may be returned | JSON/XML or other structured formats may be returned |

Example API response:

```json
{
    "city": "Delhi",
    "temperature": 32,
    "humidity": 65
}

### Real-world example

Suppose you are building a weather analytics project.

Your Python program needs today's weather:

**`Your Python Program -> API -> Weather Service -> Weather Data -> API -> Your Python Program`**

Your Python program doesn't need to know how the weather company's internal database works.

It simply asks the API: "Give me the weather data for Delhi."

The API sends the data back.

### Request and Response

This is the fundamental API workflow.

#### Request = Your program asks the server for something.

**`Example: "Give me weather data for Delhi."`**

#### Response = The server sends data back.

Example:
{
    "city": "Delhi",
    "temperature": 32,
    "humidity": 65
}

### **`So remember:`**
REQUEST  →  Ask for something

RESPONSE →  Receive something
