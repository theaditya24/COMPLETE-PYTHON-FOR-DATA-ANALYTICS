### Part 2 — Data Cleaning

```python
df.isnull().sum()
```

### Note : From here we can see we have only two columns which have missing values in them

```python
df['Rating'] = df['Rating'].fillna(df['Rating'].mean())
df['Rating'].isna().sum()
```

```python
df["Delivery_Days"] = df["Delivery_Days"].fillna(df["Delivery_Days"].median())
```

```python
# Remove Duplicate Orders
df[df.duplicated()]
```

```python
df.drop_duplicates(inplace=True)
```

```python
df[df.duplicated()]
```

```python
# Convert Order_Dte into Date_time
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
```

```python
# Create Month Column
df["Month"] = df["Order_Date"].dt.month_name()
```

```python
# Create Year Column
df['Year'] = df['Order_Date'].dt.year
```

### Part 10 — Time Series

#### Monthly Revenue

```python
monthly = df.groupby(df["Order_Date"].dt.to_period("M"))["Final_Price"].sum()

plt.plot(monthly.index.astype(str), monthly.values, marker="o")
plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

### Monthly Profit

```python
monthly_profit = df.groupby(df["Order_Date"].dt.to_period("M"))["Profit"].sum()

plt.plot(monthly_profit.index.astype(str), monthly_profit.values, marker="o")
plt.title("Monthly Profit")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.xticks(rotation=45)
plt.show()
```

### Monthly Orders

```python
orders = df.groupby(df["Order_Date"].dt.to_period("M"))["Order_ID"].count()

plt.figure(figsize=(10,5))
plt.plot(orders.index.astype(str), orders.values, marker="o")
plt.title("Monthly Orders")
plt.xlabel("Month")
plt.ylabel("Orders")
plt.xticks(rotation=45)
plt.show()
```

### Business Question 1:  Which city contributes the most revenue but also has the highest return rate?
### Answer - **`Bangalore`** generates the highest revenue (**`₹453,621.20`**), while **`Pune`** has the highest return rate (**`17.14%`**). Therefore, the city contributing the most revenue is **`not`** the city with the highest return rate.

### Business Question 2: Which product category generates the highest revenue but not the highest profit?
### **Answer:** The **`Books`** category generates the highest revenue (**`₹467,007.95`**) and also records the highest profit (**`₹132,118.84`**). Therefore, **`no product category`** generates the highest revenue without also generating the highest profit.


### Business Question 3: Does offering larger discounts lead to increased sales volume or simply reduce profit?
### **Answer:** Larger discounts do **`not`** significantly increase sales volume (Correlation = **`-0.058`**). Instead, they have a **`negative impact on profit`** (Correlation = **`-0.157`**), indicating that higher discounts mainly reduce profitability.

### Business Question 4: Which marketing channel delivers the best return on investment?
### **Answer:** The **`Referral`** marketing channel delivers the highest Return on Investment (ROI) at **`27.66%`**, making it the most effective marketing channel for generating profit.

### Business Question 5: Do Gold members purchase more frequently and generate higher profit?
### **Answer:** **`No.`** Gold members neither purchase most frequently nor generate the highest profit. **`Silver members`** purchase more frequently (**`1.66 orders per `**) and generate the highest total profit (**`₹198,791.93`**).

### Business Question 6: Is there a relationship between delivery delays and customer ratings?
### **Answer:** There is **`no meaningful relationship`** between delivery delays and customer ratings. The correlation coefficient is **`0.015`**, indicating that delivery delays have little to no impact on customer ratings.
