# CRUD OPERATION PROJECT

```python
from pathlib import Path
import os
import time

def list_files():
    """List all files and directories in the current folder"""
    path = Path('.')
    items = list(path.iterdir())  # Use iterdir() for top-level only (cleaner)
    
    print("\n" + "="*60)
    print(f"📁 Current Directory: {Path.cwd()}")
    print("="*60)
    
    if not items:
        print("No files or folders found.")
        return []
    
    files = []
    dirs = []
    
    for item in sorted(items):
        if item.is_dir():
            dirs.append(item.name)
        else:
            files.append(item.name)
    
    if dirs:
        print("\n📂 Directories:")
        for i, d in enumerate(dirs, 1):
            print(f"   {i:2}. 📁 {d}")
    
    if files:
        print("\n📄 Files:")
        for i, f in enumerate(files, len(dirs) + 1):
            print(f"   {i:2}. 📝 {f}")
    
    return [item.name for item in sorted(items)]

def create_file():
    """Create a new file"""
    list_files()
    print("\n" + "-"*40)
    name = input("Enter the name of the new file (e.g., notes.txt): ").strip()
    
    if not name:
        print("❌ File name cannot be empty!")
        return
    
    p = Path(name)
    
    if p.exists():
        print("⚠️  File already exists!")
        overwrite = input("Do you want to overwrite it? (y/n): ").strip().lower()
        if overwrite != 'y':
            return
    try:
        content = input("\nEnter the content for the file (press Enter when done):\n")
        p.write_text(content, encoding='utf-8')
        print(f"✅ File '{name}' created successfully!")
    except Exception as e:
        print(f"❌ Error creating file: {e}")

def read_file():
    """Read content of a file"""
    files = list_files()
    if not files:
        print("No files to read!")
        return
    
    print("\n" + "-"*40)
    try:
        choice = int(input("Enter the number of the file you want to read: "))
        if choice < 1 or choice > len(files):
            print("❌ Invalid choice!")
            return
        
        filename = files[choice - 1]
        p = Path(filename)
        
        if p.is_dir():
            print("❌ Cannot read a directory!")
            return
        
        content = p.read_text(encoding='utf-8')
        print("\n" + "="*60)
        print(f"📖 Content of '{filename}':")
        print("="*60)
        print(content)
        print("="*60)
        
    except ValueError:
        print("❌ Please enter a valid number!")
    except Exception as e:
        print(f"❌ Error reading file: {e}")

def update_file():
    """Update (append or overwrite) a file"""
    files = list_files()
    if not files:
        print("No files to update!")
        return
    
    print("\n" + "-"*40)
    try:
        choice = int(input("Enter the number of the file you want to update: "))
        if choice < 1 or choice > len(files):
            print("❌ Invalid choice!")
            return
        
        filename = files[choice - 1]
        p = Path(filename)
        
        if p.is_dir():
            print("❌ Cannot update a directory!")
            return
        
        print("\n1. Append content (add to existing)")
        print("2. Overwrite entire content")
        mode_choice = input("Choose (1/2): ").strip()
        
        new_content = input("\nEnter the new content:\n")
        
        if mode_choice == '1':
            with p.open('a', encoding='utf-8') as f:
                f.write("\n" + new_content)
            print(f"✅ Content appended to '{filename}' successfully!")
        elif mode_choice == '2':
            p.write_text(new_content, encoding='utf-8')
            print(f"✅ File '{filename}' overwritten successfully!")
        else:
            print("❌ Invalid option!")
            
    except ValueError:
        print("❌ Please enter a valid number!")
    except Exception as e:
        print(f"❌ Error updating file: {e}")

def delete_file():
    """Delete a file"""
    files = list_files()
    if not files:
        print("No files to delete!")
        return
    
    print("\n" + "-"*40)
    try:
        choice = int(input("Enter the number of the file you want to delete: "))
        if choice < 1 or choice > len(files):
            print("❌ Invalid choice!")
            return
        
        filename = files[choice - 1]
        p = Path(filename)
        
        if p.is_dir():
            confirm = input(f"⚠️  '{filename}' is a directory. Delete it? (y/n): ").strip().lower()
            if confirm != 'y':
                return
            import shutil
            shutil.rmtree(p)
            print(f"✅ Directory '{filename}' deleted!")
        else:
            confirm = input(f"⚠️  Are you sure you want to delete '{filename}'? (y/n): ").strip().lower()
            if confirm != 'y':
                print("Operation cancelled.")
                return
            p.unlink()
            print(f"✅ File '{filename}' deleted successfully!")
            
    except ValueError:
        print("❌ Please enter a valid number!")
    except Exception as e:
        print(f"❌ Error deleting file: {e}")

def main():
    """Main program loop"""
    print("🚀 Welcome to File CRUD Manager! 🚀")
    print("A simple interactive file management tool\n")
    
    while True:
        print("\n" + "="*60)
        print("📋 MAIN MENU")
        print("="*60)
        print("1. 📝 Create a new file")
        print("2. 📖 Read a file")
        print("3. ✏️  Update a file")
        print("4. 🗑️  Delete a file")
        print("5. 📂 List files again")
        print("6. ❌ Exit")
        print("="*60)
        
        try:
            choice = int(input("\nEnter your choice (1-6): ").strip())
            
            if choice == 1:
                create_file()
            elif choice == 2:
                read_file()
            elif choice == 3:
                update_file()
            elif choice == 4:
                delete_file()
            elif choice == 5:
                list_files()
            elif choice == 6:
                print("\n👋 Thank you for using File CRUD Manager! Goodbye!")
                time.sleep(1)
                break
            else:
                print("❌ Invalid choice! Please select 1-6.")
                
            input("\nPress Enter to continue...")
            
        except ValueError:
            print("❌ Please enter a valid number!")
            input("\nPress Enter to continue...")
        except KeyboardInterrupt:
            print("\n\n👋 Program terminated by user.")
            break

if __name__ == "__main__":
    main()
```
