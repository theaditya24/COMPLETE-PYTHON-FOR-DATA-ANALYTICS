## LABELING AXES 

### This will be now COMPLETE BASIC GRAPH

```python
time = [1, 2, 3, 4, 5]
dist = [10, 20, 30, 40, 50]

plt.plot(dist, time)

plt.title("Distance-Time Graph")

plt.xlabel("Distance")
plt.ylabel("Time")

plt.show()
```

## CHANGING LINE COLOR

```python
time = [1, 2, 3, 4, 5]
dist = [10, 20, 30, 40, 50]

plt.plot(dist, time, color='orange')

plt.title("Distance-Time Graph")

plt.xlabel("Distance")
plt.ylabel("Time")

plt.show()
```

## CHANGING LINE STYLE

```python
time = [1, 2, 3, 4, 5]
dist = [10, 20, 30, 40, 50]

plt.plot(dist, time, color='orange', linestyle='dashed')

plt.title("Distance-Time Graph")

plt.xlabel("Distance")
plt.ylabel("Time")

plt.show()
```

## CHANGING LINE WIDTH

```python
time = [1, 2, 3, 4, 5]
dist = [10, 20, 30, 40, 50]

plt.plot(dist, time, color='orange', linestyle='dotted', linewidth = 3)

plt.title("Distance-Time Graph")

plt.xlabel("Distance")
plt.ylabel("Time")

plt.show()
```

## ADDING MARKERS Markers represent actual data points.

### Now this is a Fully CUSTOMIZED LINE GRAPH

```python
time = [1, 2, 3, 4, 5]
dist = [10, 20, 30, 40, 50]

plt.plot(dist, time, color='orange', linestyle='solid', linewidth = 2, marker='o',markersize=10)

plt.title("Distance-Time Graph")

plt.xlabel("Distance")
plt.ylabel("Time")

plt.show()

# COMMON MARKERS
# 'o'  circle
# 's'  square
# '^'  triangle
# '*'  star
# '+'  plus
# 'x'  x-mark
# 'D'  diamond
```

## FIGURE SIZE

```python
time = [1, 2, 3, 4, 5]
dist = [10, 20, 30, 40, 50]

plt.figure(figsize=(10,5)) # 'width of graph = 10 inch' and 'height of graph = 5 inch'

plt.plot(dist, time, color='orange', linestyle='solid', linewidth = 2, marker='o',markersize=10)

plt.title("Distance-Time Graph")

plt.xlabel("Distance")
plt.ylabel("Time")

plt.show()

# EXPLANATION
# figsize=(width,height)  Measured in inches.
```

## GRID

```python
time = [1, 2, 3, 4, 5]
dist = [10, 20, 30, 40, 50]

plt.figure(figsize=(10,5)) # 'width of graph = 10 inch' and 'height of graph = 5 inch'

plt.plot(dist, time, color='orange', linestyle='solid', linewidth = 2, marker='o',markersize=10)

plt.grid()

plt.title("Distance-Time Graph")

plt.xlabel("Distance")
plt.ylabel("Time")

plt.show()
```

## TYPES OF GRAPHS IN MATPLOTLIB

## 1. LINE PLOT

## MULTIPLE LINES

## HORIZONTAL BAR CHART

## 3. HISTOGRAM

## OBJECT ORIENTED APPROACH (VERY IMPORTANT)

```python
fig = plt.figure()
# WHAT IS fig?
# fig = Entire figure/window
# we have just created a canvas and stored it into fig
```

```python
fig = plt.figure()
axis1 = fig.add_axes([0.1,0.1,0.9,0.9])
# add_axes([ allignment from LEFT, allignment from BOTTOM, WIDHT %, HEIGHT %]) 

x = [1,2,3,4]
y = [10,20,30,40]
axis1.plot(x,y)

axis2 = fig.add_axes([0.2,0.6,0.3,0.3]) 
axis2.plot(x,y)
plt.show()

# plt.figure() creates a blank figure/canvas.

# fig.add_axes([left, bottom, width, height])
# adds custom plotting areas (axes) inside the figure.
# Values are in percentage form from 0 to 1.

# axis1 creates the first plot area starting from:
# 10% left, 10% bottom, with 60% width and 60% height.

# axis2 creates another plot area:
# 20% left, 20% bottom, with 80% width and 80% height.

# Both axes plot the same x and y data.

# Since axis2 is larger and added later,
# it overlaps axis1 on the same figure.

# plt.show() finally displays the complete figure.
```

```python
fig, ax = plt.subplots()  # ax = Actual plotting area

x = [1,2,3]
y = [10,20,30]

ax.plot(x,y)

ax.set_title("Graph")

plt.show()
```

## MULTIPLE SUBPLOTS USING OOP

## MATPLOTLIB CUSTOM STYLING EXAMPLE

```python
# ==============================
# MATPLOTLIB LINE STYLING
# ==============================

import numpy as np
import matplotlib.pyplot as plt

# Creates 11 equally spaced values from 0 to 5
xaxis = np.linspace(0,5,11)

# Create figure and axes
fig, ax = plt.subplots(figsize=(12,6))


# =========================================================
# LINE WIDTH EXAMPLES
# =========================================================
# linewidth or lw controls thickness of line

ax.plot(xaxis, xaxis+1, color="red", linewidth=0.25)
ax.plot(xaxis, xaxis+2, color="red", linewidth=0.50)
ax.plot(xaxis, xaxis+3, color="red", linewidth=1.00)
ax.plot(xaxis, xaxis+4, color="red", linewidth=2.00)


# =========================================================
# LINE STYLE EXAMPLES
# =========================================================
# linestyle or ls controls line pattern

# '-'   = solid line
# '--'  = dashed line
# '-.'  = dash-dot line
# ':'   = dotted line

ax.plot(xaxis, xaxis+5, color="green", lw=3, linestyle='-')
ax.plot(xaxis, xaxis+6, color="green", lw=3, ls='--')
ax.plot(xaxis, xaxis+7, color="green", lw=3, ls=':')


# =========================================================
# CUSTOM DASH STYLE
# =========================================================
# set_dashes([line_length, space_length, ...])

line, = ax.plot(xaxis, xaxis+8, color="black", lw=1.50)

line.set_dashes([5, 5, 15, 5])


# =========================================================
# MARKER EXAMPLES
# =========================================================
# marker adds symbols on data points

# 'o' = circle
# '+' = plus
# 's' = square
# '1' = tri_down

ax.plot(xaxis, xaxis+9,  color="blue", lw=3, ls='--', marker='+')

ax.plot(xaxis, xaxis+10, color="blue", lw=3, ls='--', marker='o')

ax.plot(xaxis, xaxis+11, color="blue", lw=3, ls='--', marker='s')

ax.plot(xaxis, xaxis+12, color="blue", lw=3, ls='--', marker='1')


# =========================================================
# MARKER SIZE EXAMPLES
# =========================================================
# markersize controls marker size

ax.plot(xaxis, xaxis+13,
        color="purple",
        lw=1,
        ls='--',
        marker='o',
        markersize=2)

ax.plot(xaxis, xaxis+14,
        color="purple",
        lw=1,
        ls='--',
        marker='o',
        markersize=4)

ax.plot(xaxis, xaxis+15,
        color="purple",
        lw=1,
        ls='--',
        marker='o',
        markersize=8,
        markerfacecolor='red')


# =========================================================
# MARKER BORDER CUSTOMIZATION
# =========================================================
# markerfacecolor = inside color
# markeredgecolor = border color
# markeredgewidth = border thickness

ax.plot(xaxis, xaxis+16,
        color="purple",
        lw=1,
        ls='--',
        marker='s',
        markersize=8,
        markerfacecolor='yellow',
        markeredgewidth=3,
        markeredgecolor='green')


# =========================================================
# FINAL DISPLAY
# =========================================================

plt.show()
```

## ANNOTATIONS
