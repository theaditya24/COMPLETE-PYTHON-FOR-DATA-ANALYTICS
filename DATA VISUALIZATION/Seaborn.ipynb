## CATEGORICAL PLOT

### In Seaborn, Categorical Plots are used to visualize relationships where one or both variables are categorical.

### What is a Categorical Variable?

### Types of Categorical Plots in Seaborn

### 1) Bar Plot

```python
# Shows aggregated values (usually mean).
# In barplot for any one axis (x or y) you will give Categorical Data and for another axis you will give numerical Data
sns.barplot(
    x="day",
    y="total_bill",
    data=tips
)

plt.show()

# Important
# By default:
#     barplot calculates mean
#     Mean Visualization = Height of bar = average value
```

### 2) Count Plot 

```python
# Counts frequency of categories.
# COUNT PLOT usually needs only x-axis and it automatically calcultes the second one
# It basically count any category you want in the table

sns.countplot(
    x="day",
    data=tips
)

# Purpose
# Shows: how many entries exist in each category
```

### Difference Between barplot and countplot

### 3) Box Plot

```python
# Shows statistical summary of data.
sns.boxplot(
    x="day",
    y="total_bill",
    data=tips
)

plt.show()
```

### 4) Violin Plot

```python
# Combination of:
#     boxplot
#     distribution plot

sns.violinplot(
    x="day",
    y="total_bill",
    data=tips
)

plt.show()
```

### Boxplot vs Violinplot

```python
# Boxplot Components:
# -------------------
# Middle line  = Median
# Box edges    = Q1 and Q3
# Whiskers     = Range
# Dots outside = Outliers
```

### 5. STRIPPLOT

```python
# USED FOR: Showing individual data points
sns.stripplot(
    data=tips,
    x="day",
    y="total_bill",
    jitter=True
)

plt.show()
```

### 6. SWARMPLOT

```python
# BETTER VERSION OF STRIPPLOT
sns.swarmplot(
    data=tips,
    x="day",
    y="total_bill"
)

plt.show()
```

## DISTRIBUTION PLOTS

### SIMPLE UNDERSTANDING

### MAIN DISTRIBUTION PLOTS IN SEABORN

### 1) HISTOGRAM

```python
tips = sns.load_dataset("tips")

sns.histplot(
    x="total_bill",
    data=tips,
    bins=10
)

plt.show()

# What Happens Here?
    # X-axis → bill values
    # Y-axis → frequency/count

# Taller bar means: more values in that range
```

### HISTOGRAM WITH KDE

```python
# What is KDE?
# KDE means: Kernel Density Estimation
# It creates: smooth probability density curve instead of block bars.

sns.histplot(
    x="total_bill",
    kde=True,
    data=tips
)

plt.show()
```

### MULTIPLE HISTOGRAMS USING HUE

```python
sns.histplot(
    x="total_bill",
    hue="sex",
    data=tips
)
```

### COMMON PARAMETERS OF HISTPLOT

```python
sns.histplot(
    x="total_bill",
    hue="sex",
    multiple="dodge",
    data=tips
)

# TYPES OF multiple
# | Value     | Meaning      |
# | --------- | ------------ |
# | `"layer"` | Overlap      |
# | `"stack"` | Stack bars   |
# | `"dodge"` | Side-by-side |
```

### 2) KDE PLOT

```python
sns.kdeplot(
    x="total_bill",
    data=tips
)

plt.show()
```

### KDE vs Histogram

### FILLED KDE

```python
sns.kdeplot(
    x="total_bill",
    fill=True,
    data=tips
)
```

### MULTIPLE KDE PLOTS

```python
sns.kdeplot(
    x="total_bill",
    hue="sex",
    fill=True,
    data=tips
)
```

### KDE FOR 2 VARIABLES

```python
sns.kdeplot(
    x="total_bill",
    y="tip",
    fill=True,
    data=tips
)
```

### 3) RUG PLOT

```python
sns.rugplot(
    x="total_bill",
    data=tips
)

plt.show()
```

### COMBINING RUG WITH KDE

```python
sns.kdeplot(
    x="total_bill",
    fill=True,
    data=tips
)

sns.rugplot(
    x="total_bill",
    data=tips,
    height=0.08
)

plt.show()
```

## REGRESSION PLOTS

### 1. REGPLOT

```python
sns.regplot(
    data=tips,
    x="total_bill",
    y="tip",
    marker='+', # CHANGING MARKER STYLE
    ci=None,  # REMOVING CONFIDENCE INTERVAL
    color = 'green'
)

plt.show()
```

### 2. LMPLOT

```python
sns.lmplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.show()
```

```python
# FACETING USING COL
sns.lmplot(
    data=tips,
    x="total_bill",
    y="tip",
    col="sex"
)

plt.show()
```

### 3. JOINTPLOT

```python
# USED FOR: Combined scatter + distribution
sns.jointplot(
    data=tips,
    x="total_bill",
    y="tip"
)

plt.show()
```

### 4. PAIRPLOT

```python
# USED FOR: Relationship between all numeric columns
iris = sns.load_dataset("iris")
sns.pairplot(
    iris,
    hue="species"
)

plt.show()
```

## REAL-WORLD DATA ANALYSIS FLOW

```python
# STEP 1 → Load Data
# df = pd.read_csv("data.csv")
df = sns.load_dataset("tips")

# STEP 2 → Understand Data
df.head(3)
df.info()

# STEP 3 → Distribution Analysis
sns.histplot(df["total_bill"])

# STEP 4 → Relationship Analysis
sns.scatterplot(data=df, x="total_bill",y="tip")

# STEP 5 → Correlation Analysis
sns.heatmap(df.corr(numeric_only=True), annot=True)

# STEP 6 → Categorical Analysis
sns.boxplot(data=df, x="category", y="value")

# STEP 7 → Final Styling
plt.title("Analysis")
plt.show()
```

## DIFFERENCE BETWEEN MATPLOTLIB AND SEABORN

```python
# Matplotlib
# ------------
# Low-level library
# More control
# More code needed

# Seaborn
# ----------
# High-level library
# Beautiful defaults
# Less code
# Better statistical plots

# IMPORTANT:
# Seaborn internally uses Matplotlib
```

## MOST IMPORTANT INTERVIEW QUESTIONS

### Q1. Why use Seaborn instead of Matplotlib?

```python
# Because:
# 1. Better visualization
# 2. Less code
# 3. Statistical plotting
# 4. Better DataFrame support
```

### Q2. Difference between countplot and barplot?

```python
# countplot:
# Counts category frequency

# barplot:
# Shows aggregated values (mean by default)
```

### Q3. Difference between stripplot and swarmplot?

```python
# stripplot:
# Points may overlap

# swarmplot:
# Adjusts points to avoid overlap
```

### Q4. Difference between regplot and lmplot?

```python
# regplot:
# Axes-level function

# lmplot:
# Figure-level function
# Supports faceting
```

## COMPLETE MINI PROJECT

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

# Set theme
sns.set_style("darkgrid")

# Create figure
plt.figure(figsize=(10,5))

# Plot
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    size="size"
)

# Labels
plt.title("Restaurant Tips Analysis")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

# Show
plt.show()
```

## ANALYSIS OVER THIS GRAPH

### 1. WHAT DOES THIS GRAPH REPRESENT?

```python
# X-axis  -> Total restaurant bill
# Y-axis  -> Tip amount given
# Each dot represents one customer entry/order.

# So every point means:
    # One restaurant bill
    # One tip amount
    # One customer gender
    # One group size
```

### 2. COLOR ANALYSIS (HUE = SEX)

```python
# Blue   -> Male
# Orange -> Female
```

### 3. SIZE ANALYSIS (SIZE = SIZE)

```python
# The circle size represents: Number of people at the table/group

# Small circles:
    # Smaller group size
    # Example: 1 or 2 people
    
# Bigger circles:
    # Larger group size
    # Example: 5 or 6 people
```

## 4. MAIN RELATIONSHIP IN THE GRAPH
## IMPORTANT OBSERVATION

## WHAT DOES POSITIVE RELATIONSHIP MEAN?

```python
# As total bill increases, tip amount also generally increases.

# You can clearly see:
    # Lower bills -> lower tips
    # Higher bills -> higher tips
```

## CONCLUSION

```python
# Customers who spend more money
# generally give higher tips.
```

## 5. DATA CLUSTER ANALYSIS

```python
# Most points are concentrated around:
    # Total Bill = 10 to 25
    # Tip = 2 to 4

# Meaning:
    # Most restaurant customers spend
    # between 10 and 25 dollars
    # and give tips between 2 and 4 dollars.
```

## 6. OUTLIER ANALYSIS
## OUTLIERS = Unusual points

```python
# Total bill around 50
# Tip around 10

# This is much higher than normal.

# Meaning:
# Some customers gave exceptionally high tips.
```

## 7. GENDER-BASED ANALYSIS

```python
# Observation:
# Male customers have more high-bill entries.

# You can see many blue points at:
# Total bill > 30
    
# Another Observation
# Male customers also appear to give
# some of the highest tips.

# BUT IMPORTANT NOTE

# This DOES NOT mean: All males tip more.

# Because: Dataset may contain more male customers overall.
# So we should not make unfair conclusions.
```
