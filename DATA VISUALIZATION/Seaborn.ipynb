```python
emp_df = pd.DataFrame({
    'Employee': ['Aman', 'Riya', 'Karan', 'Neha', 'Rahul', 'Priya'],
    'Experience': [1, 2, 3, 4, 5, 6],
    'Salary': [30000, 35000, 42000, 50000, 58000, 68000],
    'Department': ['IT', 'HR', 'IT', 'Sales', 'HR', 'IT']
})

emp_df
```

### Does salary tend to increase with employee experience?

```python
sns.scatterplot(data=emp_df, x='Experience', y='Salary')
plt.title('Experience vs Salary')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')

plt.show()
```

**`data=df`** tells Seaborn: "The dataset I want you to use is stored in the DataFrame called df."

**`x='Experience'`** means: Take the Experience column from df and place it on the horizontal X-axis.

**`y='Salary'`** means: Take the Salary column from df and represent it on the vertical Y-axis.

#### We can see that as Experience increases:
    Experience ↑
    Salary ↑
    
    So a reasonable interpretation is: Employees with greater experience generally have higher salaries in this dataset.

#### Business insights
Now imagine this is real company data.
We could investigate conclusions such as:

    - Insight 1
    Salary appears to increase with experience.
    This suggests the company may reward experience through higher compensation.
            
    - Insight 2
    If an employee has high experience but unusually low salary, that employee might be worth investigating for potential pay inequity.

```python
sales_df = pd.DataFrame({
    'Product': ['Laptop', 'Shoes', 'Phone', 'T-Shirt', 'Watch',
                'Headphones', 'Jeans', 'Tablet', 'Jacket', 'Smartwatch'],
    
    'Category': ['Electronics', 'Fashion', 'Electronics', 'Fashion', 'Accessories',
                 'Electronics', 'Fashion', 'Electronics', 'Fashion', 'Accessories'],
    
    'Price': [65000, 3000, 30000, 1000, 5000,
              4000, 2500, 25000, 4500, 12000],
    
    'Rating': [4.5, 4.2, 4.6, 3.9, 4.1,
               4.4, 4.0, 4.3, 4.2, 4.5],
    
    'Sales': [120, 450, 300, 600, 250,
              420, 500, 280, 350, 220]
})

sales_df
```

### Do more expensive products generally have lower or higher sales?

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=sales_df,
    x='Price',
    y='Sales'
)

plt.title('Relationship Between Product Price and Sales')
plt.xlabel('Product Price (₹)')
plt.ylabel('Sales')

plt.show()
```

#### Bivariate Analysis: Price vs Sales

#### Business Question
Do more expensive products generally have lower or higher sales?

#### Variables
- **`Price:`** Numerical
- **`Sales:`** Numerical

#### Type of Analysis
This is **`Bivariate Analysis`** because we are analyzing the relationship between two variables.

#### Visualization Used
A **`Scatter Plot`** is used because both Price and Sales are numerical variables, and we want to examine the relationship between them.

#### Interpretation
The scatter plot shows that products with higher prices generally tend to have lower sales, while lower-priced products tend to have higher sales.

#### Business Insight
This suggests that customers may be price-sensitive. Lower-priced products appear to sell more units, while expensive products tend to have lower sales volume.

However, price alone may not explain sales. Product category, customer demand, brand, discounts, and other factors may also influence sales.

### Does the relationship between Price and Sales differ across Product Categories?

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=sales_df,
    x='Price',
    y='Sales',
    hue='Category',palette = 'deep'
)

plt.title('Price vs Sales Across Product Categories')
plt.xlabel('Product Price (₹)')
plt.ylabel('Sales')

plt.show()
```

#### Multivariate Analysis: Price vs Sales by Product Category

#### Business Question
Does the relationship between product price and sales differ across product categories?

#### Variables
- **`Price:`** Numerical
- **`Sales:`** Numerical
- **`Category:`** Categorical

#### Type of Analysis
This is **`Multivariate Analysis`** because three variables are being analyzed together.

#### Visualization Used
A **`Scatter Plot with hue`** is used.

- X-axis represents **`Price`**
- Y-axis represents **`Sales`**
- Color (`hue`) represents **`Product Category`**

#### Interpretation
The visualization allows us to study the relationship between Price and Sales while also comparing products from different categories.

Different colors represent different product categories, making it easier to identify whether the Price-Sales relationship varies between Electronics, Fashion, and Accessories.

#### Business Insight
The company can identify whether customers in different product categories respond differently to price.

For example, some categories may maintain relatively strong sales even at higher prices, while other categories may experience lower sales as prices increase.

This information can help the business make better category-specific pricing and marketing decisions.

## What Does Each Feature Represent?

| Feature | Represents |
|---------|------------|
| **X-axis** | Total Bill |
| **Y-axis** | Tip Amount |
| **Color** | Male / Female |
| **Size** | Number of people at the table |
| **Style** | Smoker / Non-smoker |

---

## 1. Total Bill vs Tip

- The graph shows a **positive relationship** between `total_bill` and `tip`.
- As the **total bill increases, the tip generally increases**.
- Customers with smaller bills usually give smaller tips, while customers with larger bills tend to give larger tips.

### Conclusion
> **Total bill and tip have a positive correlation.**

**Important:** Correlation does not mean causation. We cannot say that a higher bill **causes** a higher tip just from this graph.

---

## 2. Most Customers Have Moderate Bills and Tips

Most of the data points are concentrated around:

- **Total Bill:** 10–25
- **Tip:** 1.5–4

### Conclusion
> **Most customers have moderate total bills and give moderate tips.**

---

## 3. Possible Outliers

A few points are far away from the main group of observations.

For example:

- Total Bill ≈ 50 → Tip ≈ 10
- Total Bill ≈ 48 → Tip ≈ 9

### Conclusion
> **There are a few unusually high bill and tip values that could be investigated as potential outliers.**

**Important:** An outlier does not always mean incorrect or bad data. It can be a completely valid observation.

---

## 4. Male vs Female Customers

- 🔵 **Blue** → Male
- 🟠 **Orange** → Female

Both male and female customers show a similar pattern:

**Higher Bill → Generally Higher Tip**

There appear to be more male observations in the dataset, especially among some of the higher bill values.

### Conclusion
> **Both male and female customers show a positive relationship between total bill and tip.**

**Do not directly conclude:**  
> "Males give higher tips than females."

To make that conclusion, we should perform further analysis using:

- Average tip
- Median tip
- Tip percentage

---

## 5. Smokers vs Non-Smokers

The marker style represents smoking status:

- **● → Smoker**
- **× → Non-smoker**

Both smokers and non-smokers are spread throughout the graph.

There is no very clear separation between the two groups.

### Conclusion
> **Smokers and non-smokers appear to have similar general bill-tip patterns.**

Further statistical analysis would be needed to determine whether one group actually tips more than the other.

---

## 6. Table Size

The **size of each point** represents the number of people at the table.

Generally, some larger groups also have larger total bills.

This makes logical sense:

**More People → More Orders → Potentially Higher Total Bill**

### Conclusion
> **Larger groups may tend to have higher total bills.**

However, a separate analysis of `size` vs `total_bill` would give us a clearer answer.

---

# Overall Conclusion

From this scatter plot, we can mainly observe:

1. **Total bill and tip have a positive relationship.**
2. **Most customers have bills between 10–25 and tips between 1.5–4.**
3. **A few potential outliers are present.**
4. **Both males and females show a similar bill-tip relationship.**
5. **There is no obvious difference between smokers and non-smokers from this graph alone.**
6. **Larger groups may be associated with larger total bills.**

> **Main Insight:** As the total bill increases, customers generally tend to leave a higher tip.

## 2) Count Plot 

#### 📊 Count Plot (`sns.countplot()`)

#### What is a Count Plot?

A **Count Plot** displays the **frequency (count)** of each category in a **categorical variable**.

> **Key Point:** A Count Plot usually requires **only the `x` (or `y`) axis**, and Seaborn automatically calculates the count for each category.

---

#### 🎯 When Should You Use a Count Plot?

Use a Count Plot whenever your business question begins with:

- **How many...?**
- **How often...?**
- **How frequently...?**
- **Number of...?**
- **Count of...?**

---

#### 📌 Real-World Business Examples

- ✅ **How many customers belong to each city?**
- ✅ **Number of employees in each department**
- ✅ **Number of hotel bookings by room type**
- ✅ **Number of orders by payment method**
- ✅ **Number of students in each grade**
- ✅ **How many products are in each category?**
- ✅ **Number of passengers by travel class**
- ✅ **How many users signed up from each country?**

---

#### 💡 Interview Tip

> **Count Plot = Categorical Data + Frequency**

Whenever you need to **count the occurrences of categories**, a **Count Plot** is usually the best choice.

## COMPLETE MINI PROJECT

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

# Set theme
sns.set_style("darkgrid")

# Create figure
plt.figure(figsize=(10,5))

# Plot
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="sex",
    size="size"
)

# Labels
plt.title("Restaurant Tips Analysis")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

# Show
plt.show()
```

## ANALYSIS OVER THIS GRAPH

### 1. WHAT DOES THIS GRAPH REPRESENT?

```python
# X-axis  -> Total restaurant bill
# Y-axis  -> Tip amount given
# Each dot represents one customer entry/order.

# So every point means:
    # One restaurant bill
    # One tip amount
    # One customer gender
    # One group size
```

### 2. COLOR ANALYSIS (HUE = SEX)

```python
# Blue   -> Male
# Orange -> Female
```

### 3. SIZE ANALYSIS (SIZE = SIZE)

```python
# The circle size represents: Number of people at the table/group

# Small circles:
    # Smaller group size
    # Example: 1 or 2 people
    
# Bigger circles:
    # Larger group size
    # Example: 5 or 6 people
```

## 4. MAIN RELATIONSHIP IN THE GRAPH
## IMPORTANT OBSERVATION

## WHAT DOES POSITIVE RELATIONSHIP MEAN?

```python
# As total bill increases, tip amount also generally increases.

# You can clearly see:
    # Lower bills -> lower tips
    # Higher bills -> higher tips
```

## CONCLUSION

```python
# Customers who spend more money
# generally give higher tips.
```

## 5. DATA CLUSTER ANALYSIS

```python
# Most points are concentrated around:
    # Total Bill = 10 to 25
    # Tip = 2 to 4

# Meaning:
    # Most restaurant customers spend
    # between 10 and 25 dollars
    # and give tips between 2 and 4 dollars.
```

## 6. OUTLIER ANALYSIS
## OUTLIERS = Unusual points

```python
# Total bill around 50
# Tip around 10

# This is much higher than normal.

# Meaning:
# Some customers gave exceptionally high tips.
```

## 7. GENDER-BASED ANALYSIS

```python
# Observation:
# Male customers have more high-bill entries.

# You can see many blue points at:
# Total bill > 30
    
# Another Observation
# Male customers also appear to give
# some of the highest tips.

# BUT IMPORTANT NOTE

# This DOES NOT mean: All males tip more.

# Because: Dataset may contain more male customers overall.
# So we should not make unfair conclusions.
```
