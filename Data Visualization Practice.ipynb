```python
np.random.seed(10)
innings_one = pd.DataFrame({'Overs':  np.arange(1,21),
                            'Runs': np.random.randint(0,37,20)})
innings_two = pd.DataFrame({'Overs':  np.arange(1,21),
                            'Runs': np.random.randint(0,37,20)})
```

```python
plt.figure(figsize=(8,6))
plt.plot(innings_one['Overs'],innings_one['Runs'].cumsum(),label="First Innings")
plt.plot(innings_two['Overs'],innings_two['Runs'].cumsum(),label="Second Innings")
plt.xticks(range(1, 21))   # Show every over from 1 to 20
plt.grid()
plt.title("Innings 1 vs Innings 2")
plt.xlabel("Overs")
plt.ylabel("Runs")
plt.legend()  # Legend helps identify multiple lines.

plt.show()
```

```python
overs = np.arange(1, 41)

plt.figure(figsize=(12,6))

plt.plot(overs, pd.concat([innings_one['Runs'],innings_two['Runs']], ignore_index=True).cumsum(), marker='o')

plt.xticks(range(1,41))
plt.xlabel("Overs")
plt.ylabel("Total Runs Scored")
plt.title("Total Runs Scored Throughout the 40-Over Match")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(12,6))

bar_width = 0.4

plt.bar(innings_one['Overs'] - bar_width/2,
        innings_one['Runs'],
        width=bar_width,
        label='First Innings')

plt.bar(innings_two['Overs'] + bar_width/2,
        innings_two['Runs'],
        width=bar_width,
        label='Second Innings')

plt.xticks(range(1,21))
plt.xlabel("Overs")
plt.ylabel("Runs Scored")
plt.title("Runs Per Over")
plt.grid(axis='y')
plt.legend()

plt.show()
```

```python
import pandas as pd

df = pd.DataFrame({
    "Date": pd.date_range(start="2025-01-01", periods=30),
    "Region": [
        "North","South","East","West","North","South","East","West","North","South",
        "East","West","North","South","East","West","North","South","East","West",
        "North","South","East","West","North","South","East","West","North","South"
    ],
    "Category": [
        "Electronics","Fashion","Grocery","Electronics","Furniture","Fashion","Grocery","Electronics",
        "Furniture","Fashion","Grocery","Electronics","Furniture","Fashion","Grocery","Electronics",
        "Furniture","Fashion","Grocery","Electronics","Furniture","Fashion","Grocery","Electronics",
        "Furniture","Fashion","Grocery","Electronics","Furniture","Fashion"
    ],
    "Orders": [
        120,95,150,110,80,130,170,115,90,140,
        165,105,75,135,180,125,85,145,175,130,
        95,155,185,120,100,160,190,135,105,170
    ],
    "Revenue": [
        240000,180000,210000,260000,170000,220000,195000,255000,185000,230000,
        205000,250000,175000,225000,200000,265000,180000,235000,210000,275000,
        190000,240000,215000,280000,195000,245000,220000,285000,200000,250000
    ],
    "Profit": [
        42000,32000,39000,47000,28000,41000,36000,49000,30000,43000,
        38000,50000,29000,42000,37000,51000,31000,44000,39000,53000,
        33000,45000,40000,55000,34000,46000,41000,56000,35000,47000
    ],
    "Customer_Rating": [
        4.3,4.1,4.5,4.4,3.9,4.2,4.6,4.5,4.0,4.3,
        4.7,4.4,3.8,4.2,4.6,4.5,3.9,4.3,4.7,4.6,
        4.1,4.4,4.8,4.7,4.0,4.5,4.9,4.8,4.2,4.6
    ]
})

df
```

```python
plt.figure(figsize=(30,20))

# Question 1 
plt.subplot(2,3,1)
plt.plot(df['Date'],df['Revenue'],marker='o',markerfacecolor='red')
plt.xticks(rotation=45)
plt.title("Revenue over Date")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.grid()

# Question 2
plt.subplot(2,3,2)
plt.plot(df['Date'],df['Orders'], marker='o',markerfacecolor='red')
plt.xticks(rotation=45)
plt.title("Orders over Date")
plt.xlabel("Date")
plt.ylabel("Orders")
plt.grid()

# Question 3
plt.subplot(2,3,3)
plt.plot(df['Date'],df['Profit'], marker='o',markerfacecolor='red')
plt.xticks(rotation=45)
plt.title("Profit over Date")
plt.xlabel("Date")
plt.ylabel("Profit")
plt.grid()

# Question 4
plt.subplot(2,3,4)
plt.plot(df['Date'],df['Customer_Rating'], marker='o',markerfacecolor='red')
plt.xticks(rotation=45)
plt.title("Customer Rating over Date")
plt.xlabel("Date")
plt.ylabel("Customer Rating")
plt.grid()

# Question 5
plt.subplot(2,3,5)
plt.plot(df['Date'],df['Revenue'], label="revenue")
plt.plot(df['Date'],df['Profit'], label="profit")
plt.xticks(rotation=45)
plt.legend()
plt.title("Revenue vs Profit Over Date")
plt.xlabel("Date")
plt.grid()

# Question 6
plt.subplot(2,3,6)
north = df[df['Region'] == 'North']
south = df[df['Region'] == 'South']
plt.plot(north['Date'], north['Revenue'], label='North')
plt.plot(south['Date'], south['Revenue'], label='South')
plt.xticks(rotation=45)
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.legend()
plt.grid()

plt.tight_layout()   
```

```python
plt.figure(figsize=(20,10))

# Question 7
plt.subplot(2,3,1)
electronics = df[df['Category'] == 'Electronics']
fashion = df[df['Category'] == 'Fashion']
plt.plot(electronics['Date'], electronics['Revenue'], label='Electronics')
plt.plot(fashion['Date'], fashion['Revenue'], label='Fashion')
plt.legend()
plt.grid()
plt.xticks(rotation=45)
plt.title("Electronics Revenue vs Fashion Revenue OverTime")
plt.xlabel("Date")
plt.ylabel("Revenue")

# Question 8
plt.subplot(2,3,2)
plt.plot(df['Date'], df['Revenue'].cumsum())
plt.title("Cumulative Revenue")
plt.xticks(rotation=45)
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.grid()

# Question 9
plt.subplot(2,3,3)
plt.plot(df['Date'], df['Revenue'].rolling(window=7).mean())
plt.title("7-Day Rolling Average of Revenue")
plt.xticks(rotation=45)
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.grid()

# Question 10
plt.subplot(2,3,4)
plt.plot(df['Date'], df['Revenue'])
max_index = df['Revenue'].idxmax() # this will give us the index at which value of revenue is maximum
plt.scatter(df.loc[max_index, 'Date'],
            df.loc[max_index, 'Revenue'],
            color='red',
            s=100)
plt.title("Highest Revenue Point")
plt.xticks(rotation=45)
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.grid()

# Question 11
plt.subplot(2,3,5)
plt.plot(df['Date'], df['Revenue'], label='Revenue')
plt.plot(df['Date'], df['Profit'], label='Profit')
plt.legend(loc='upper left')
plt.title("Legend in the Upper Left")
plt.xticks(rotation=45)
plt.xlabel("Date")
plt.grid()


plt.tight_layout()   # Adjust spacing between subplots
plt.show()           # Only once
```

```python
plt.figure(figsize=(20,10))

# Question 12
plt.subplot(2,3,1)
plt.plot(df['Date'], df['Profit'] / df['Revenue'] * 100)
plt.xticks(rotation=45)
plt.xlabel("Date")
plt.ylabel("Daily Profit Margin")
plt.title("Daily Profit Margin Over Time")
plt.grid()

# Question 13
plt.subplot(2,3,2)
plt.plot(df['Date'], df['Revenue'], label='Revenue')

high_revenue = df[df['Revenue'] > 250000]

plt.scatter(high_revenue['Date'],
            high_revenue['Revenue'],
            color='red',
            s=100,
            label='Revenue > 250000')

plt.title("Revenue Over Time")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.grid()
plt.legend()

# Question 14
plt.subplot(2,3,3)
weekly_avg = df.resample('W', on='Date')['Revenue'].mean()
plt.plot(weekly_avg.index, weekly_avg.values, marker='o')
plt.xticks(rotation=45)
plt.title("Average Weekly Revenue")
plt.xlabel("Week")
plt.ylabel("Average Revenue")
plt.grid()

# Question 15
plt.subplot(2,3,4)
monthly_profit = df.groupby(df['Date'].dt.to_period('M'))['Profit'].sum().cumsum()
plt.plot(monthly_profit.index.astype(str),
         monthly_profit.values,
         marker='o')

plt.title("Monthly Cumulative Profit")
plt.xlabel("Month")
plt.ylabel("Profit")
plt.grid()

# Question 16
plt.subplot(2,3,5)
weekends = df[df['Date'].dt.dayofweek >= 5]

plt.plot(weekends['Date'],
         weekends['Revenue'],
         marker='o')
plt.title("Weekend Revenue")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.grid()

# Question 17 
plt.subplot(2,3,6)
first_half = df.iloc[:15]
second_half = df.iloc[15:]

plt.plot(first_half['Date'], first_half['Revenue'], label='First 15 Days')
plt.plot(second_half['Date'], second_half['Revenue'], label='Last 15 Days')

plt.title("First 15 Days vs Last 15 Days")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.grid()
plt.legend()

plt.tight_layout()
plt.show()
```
