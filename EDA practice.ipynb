```python
def parse_date(x):
    for fmt in ('%Y-%m-%d', '%d/%m/%Y', '%d-%m-%Y', '%Y.%m.%d'):
        try:
            return pd.to_datetime(x, format=fmt)
        except (ValueError, TypeError):
            continue
    return pd.to_datetime(x, errors='coerce')  # fallback for "26th Feb 2025" style

df['order_date_clean'] = df['order_date'].apply(parse_date)
df['order_date_clean'].isnull().sum()   
```

```python
df['order_date'] = df['order_date_clean']
```

```python
df['unit_price'] = (df['unit_price'].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False))

df['unit_price'] = pd.to_numeric(df['unit_price'], errors='coerce')
```

```python
df['order_total'] = (df['order_total'].astype(str).str.replace(',', '', regex=False))

df['order_total'] = pd.to_numeric(df['order_total'], errors='coerce')
```

```python
# Convert impossible ages to NaN
df.loc[
    (df['customer_age'] < 18) | (df['customer_age'] > 100),
    'customer_age'
] = np.nan

# Fill missing ages with median
median_age = df['customer_age'].median()
df['customer_age'] = df['customer_age'].fillna(median_age)
```

```python
# Final check
print("Missing after cleaning:", df['customer_age'].isnull().sum())
```
