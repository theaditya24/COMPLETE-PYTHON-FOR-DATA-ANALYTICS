#### Why is set_index() preferred?
    Cleaner
    No duplicate information:
        Emp_ID
        101 Ravi
        102 Priya
    
    instead of 
        Emp_ID
        101  101 Ravi
        102  102 Priya
        
    More options
    df.set_index('Emp_ID', drop=False)  It keeps the column.
    
    df.set_index('Emp_ID', inplace=True) modifies the DataFrame directly.
    
    df.set_index(['Department', 'Emp_ID']) creates a MultiIndex.

### How to access all rows with City = 'Delhi'?
#### Method 1: Specify both levels

```python
df.loc[(slice(None), 'Delhi'), :].reset_index()

# Here:
# slice(None) means "all departments".
# 'Delhi' means City = Delhi.
```

#### Method 2: Swap Index Levels

```python
df2 = df.swaplevel()
df2.loc['Delhi'].reset_index()
```
