# Difference Between merge() and concat()

### merge() is used to combine DataFrames based on common columns or indexes, similar to SQL joins

### whereas concat() is used to stack DataFrames either vertically or horizontally without requiring any matching key

### I use merge() when working with relational datasets and concat() when appending rows or combining columns from multiple DataFrames.

### example of merge()

```python
import pandas as pd

df1 = pd.DataFrame({
    'Emp_ID': [1, 2, 3],
    'Name': ['John', 'Alice', 'Bob']
})

df2 = pd.DataFrame({
    'Emp_ID': [1, 2, 3],
    'Salary': [50000, 60000, 55000]
})

# result = pd.merge(df1, df2, on='Emp_ID', how='inner')
# or
result = df1.merge(df2, on='Emp_ID',how='inner')
print(result)
```

### Example of concat()

```python
df1 = pd.DataFrame({
    'Name': ['John', 'Alice']
})

df2 = pd.DataFrame({
    'Name': ['Bob', 'David']
})

result = pd.concat([df1, df2], axis=0, ignore_index=True)
print(result)
```

# Explain categorical Data in Pandas

### 📌 Categorical Data in Pandas

Categorical data represents variables that contain a **limited number of distinct values (categories)**.  
In Pandas, the **`category`** data type is used to efficiently store and process such data.

---

### ✨ Examples of Categorical Data

| Variable | Categories |
|------------|------------|
| 👤 **Gender** | Male, Female |
| 🌍 **Region** | North, South, East, West |
| 💳 **Payment Method** | Cash, Card, UPI |

---

### ✅ Key Points

- Contains a **fixed set of categories**.
- Values are usually **text labels**.
- Stored efficiently using Pandas' **`category`** data type.
- Commonly used in **data analysis and machine learning**.

```python
import pandas as pd

df = pd.DataFrame({
    'Gender': ['Male', 'Female', 'Male'],
    'Region': ['North', 'South', 'East']
})

# Convert columns to categorical type
df['Gender'] = df['Gender'].astype('category')
df['Region'] = df['Region'].astype('category')

print(df.dtypes)
```

# 📌 `map()` vs `apply()` vs `applymap()` in Pandas

These functions are used to apply transformations to data, but they differ in scope and usage.

---

### 🔍 Comparison Table

| Feature | 🗺️ `map()` | ⚙️ `apply()` | 🔄 `applymap()` |
|-----------|------------|--------------|----------------|
| **Works on** | **Series only** | Series and DataFrame | DataFrame only |
| **Applies function to** | Each element in a Series | Rows, columns, or Series elements | Each element in a DataFrame |
| **Axis parameter** | ❌ No | ✅ Yes (`axis=0` or `axis=1`) | ❌ No |
| **Common use** | Value mapping, transformations | Aggregation and custom operations | Element-wise formatting/transformation |
| **Accepts dictionary** | ✅ Yes | ❌ No | ❌ No |


```python
import pandas as pd

df = pd.DataFrame({
    'Gender': ['M', 'F', 'O']
})

df['Gender'] = df['Gender'].map({
    'M': 'Male',
    'F': 'Female'
})

print(df)
# Since we haven't given any map() value for 'O' 
# Hence it will give NaN for that 
```

```python
df = pd.DataFrame({
    'A': [1,2,3],
    'B': [4,5,6]
})

print(df.apply(sum))
```

```python
df['Total'] = df.apply(lambda row: row['A'] + row['B'], axis=1)
df
```

```python
df = pd.DataFrame({
    'A': [1,2],
    'B': [3,4]
})

df = df.applymap(lambda x: x*10) # Element-wise on Entire DataFrame

print(df)
```
