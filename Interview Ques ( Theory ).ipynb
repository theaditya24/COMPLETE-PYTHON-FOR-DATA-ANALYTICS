# Interview Question 3: How Do You Handle Missing Values?

## Interview Question
**How do you handle missing values in a dataset?**

## Answer

Missing values occur due to incomplete data entry, system errors, unavailable information, or data corruption.

I follow a structured approach:

### Step 1: Identify Missing Values

```python
df.isnull().sum()
```

### Step 2: Calculate Missing Percentage

Understand how much data is missing before making a decision.

### Step 3: Understand Business Context

Determine whether the missing values are important or can be safely removed.

### Step 4: Choose the Appropriate Method

#### Remove Rows

Use when only a few records contain missing values.

```python
df.dropna()
```

#### Remove Columns

Use when a column contains a very high percentage of missing values and is not useful.

```python
df.drop(columns=['column_name'])
```

#### Fill Numerical Values

Use **Mean**
- Normally distributed data
- No significant outliers

Use **Median**
- Skewed data
- Presence of outliers

```python
df['Age'].fillna(df['Age'].median())
```

#### Fill Categorical Values

Use Mode.

```python
df['Gender'].fillna(df['Gender'].mode()[0])
```

Use **"Unknown"** when the missing value itself is meaningful.

#### Time-Series Data

Use:
- Forward Fill
- Backward Fill

### Step 5: Validate the Data

Ensure missing values have been handled correctly.

## Decision Table

| Situation | Method |
|-----------|--------|
| Few missing rows | Drop rows |
| Mostly empty column | Drop column |
| Numerical (Normal Distribution) | Mean |
| Numerical (Skewed Distribution) | Median |
| Categorical | Mode |
| Meaningful Missing | Unknown |
| Time Series | Forward Fill / Backward Fill |

## Common Follow-up Questions

- Why use Median instead of Mean?
- Why not always use dropna()?
- When should a column be dropped?
- When should "Unknown" be used?

## Interview Tip

Never say:
> "I always use Mean."

Instead say:
> "The method depends on the data distribution, percentage of missing values, and business context."

# Interview Question 4: How Do You Handle Duplicate Data?

## Interview Question
**How do you identify and handle duplicate data in a dataset?**

## Answer

Duplicate records occur when the same observation appears more than once. They can lead to incorrect analysis by inflating counts, averages, sums, and other statistics.

### Step 1: Identify Duplicates

```python
df.duplicated().sum()
```

Returns the number of duplicate rows.

To view duplicate rows:

```python
df[df.duplicated()]
```

### Step 2: Analyze Duplicates

Before removing duplicates, verify:
- Are they exact duplicates?
- Are they valid repeated transactions?
- Should only specific columns be checked?

### Step 3: Remove Duplicates

```python
df.drop_duplicates(inplace=True)
```

Using specific columns:

```python
df.drop_duplicates(subset=['Customer_ID'])
```

### Step 4: Validate

```python
df.duplicated().sum()
```

Ensure no unwanted duplicates remain.

## Why are duplicates harmful?

- Incorrect averages
- Inflated counts
- Misleading business insights
- Biased machine learning models

## Common Follow-up Questions

- Difference between duplicate rows and duplicate values?
- When should duplicates NOT be removed?
- How do you remove duplicates based on one column?

## Interview Tip

Never remove duplicates without understanding whether they represent genuine repeated business events.

# Interview Question 5: How Do You Handle Outliers?

## Interview Question
**What are outliers? How do you detect and handle them?**

## Answer

Outliers are unusually high or low values that differ significantly from the rest of the data.

Example:

Salary

30000
32000
31000
35000
850000

Here, **850000** is an outlier.

## Why do outliers matter?

- Distort Mean
- Affect statistical analysis
- Reduce ML model performance
- Produce misleading visualizations

## Detecting Outliers

### 1. Box Plot

Quick visual detection.

### 2. IQR Method

IQR = Q3 − Q1

Lower Bound = Q1 − 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values outside these limits are considered outliers.

### 3. Z-Score

Useful when data follows a normal distribution.

## Handling Outliers

- Remove them (if data entry errors)
- Cap (Winsorization)
- Transform data (Log Transformation)
- Keep them if they are genuine business values

## Example

A CEO's salary may appear as an outlier but is a valid observation and should not be removed.

## Common Follow-up Questions

- Should every outlier be removed?
- Why is Median preferred when outliers exist?
- Which visualization helps detect outliers?

## Interview Tip

Always verify whether an outlier is a data error or a valid business observation before removing it.

# Interview Question 6: Difference Between Mean, Median, and Mode

## Interview Question
**What is the difference between Mean, Median, and Mode? When would you use each?**

## Answer

### Mean

The arithmetic average of all values.

Formula:

Mean = Sum of Values / Number of Values

Best used when:
- Data is normally distributed.
- No significant outliers exist.

Example:

10, 20, 30

Mean = 20

---

### Median

The middle value after sorting the data.

Best used when:
- Data is skewed.
- Outliers are present.

Example:

10, 20, 30, 500

Median = 25

---

### Mode

The most frequently occurring value.

Best used for:
- Categorical variables
- Most common category

Example:

Red, Blue, Red, Green

Mode = Red

---

## Comparison Table

| Measure | Best For | Affected by Outliers |
|----------|----------|----------------------|
| Mean | Normal Distribution | Yes |
| Median | Skewed Distribution | No |
| Mode | Categorical Data | No |

## Common Interview Questions

- Why use Median instead of Mean?
- Can a dataset have multiple modes?
- Which measure is suitable for categorical data?

## Interview Tip

Remember:

Mean → Normal Distribution

Median → Skewed Distribution

Mode → Categories

# Interview Question 7: Difference Between Correlation and Covariance

## Interview Question
**What is the difference between Correlation and Covariance?**

## Answer

Both measure the relationship between two numerical variables.

---

## Covariance

Covariance indicates whether two variables move together.

Positive Covariance:
- One increases, the other also increases.

Negative Covariance:
- One increases, the other decreases.

Problem:
Its value depends on the units of measurement, making interpretation difficult.

---

## Correlation

Correlation measures both the **strength** and **direction** of the relationship.

Range:

-1 → Perfect Negative Correlation

0 → No Correlation

+1 → Perfect Positive Correlation

Unlike covariance, correlation is unit-free and standardized.

---

## Comparison

| Covariance | Correlation |
|------------|-------------|
| Measures direction | Measures direction + strength |
| Depends on units | Independent of units |
| No fixed range | Range is -1 to +1 |
| Hard to interpret | Easy to interpret |

## Example

Height and Weight

Covariance:
Positive value → They increase together.

Correlation:
0.92 → Strong positive relationship.

## Common Interview Questions

- Can correlation be greater than 1?
- What does correlation = 0 mean?
- Does correlation imply causation?

Answer:
No. Correlation does not imply causation.

## Interview Tip

Interviewers often ask:

"What is a strong correlation?"

Generally:

0.7 to 1.0 → Strong

0.3 to 0.7 → Moderate

0 to 0.3 → Weak

# Interview Question 9: Difference Between groupby() and pivot_table()

## Interview Question

**What is the difference between `groupby()` and `pivot_table()` in Pandas? When would you use each?**

## Answer

Both `groupby()` and `pivot_table()` are used to summarize and aggregate data, but they serve different purposes.

---

## groupby()

### Purpose

Groups data based on one or more columns and performs aggregation on each group.

### Syntax

```python
df.groupby('Department')['Salary'].mean()
```

### Example

| Department | Salary |
|------------|--------|
| IT | 50000 |
| HR | 40000 |
| IT | 60000 |
| HR | 45000 |

```python
df.groupby('Department')['Salary'].mean()
```

Output:

| Department | Average Salary |
|------------|----------------|
| HR | 42500 |
| IT | 55000 |

### Features

- Groups data based on one or more columns.
- Supports multiple aggregation functions.
- Returns a Series or DataFrame.
- Highly flexible for complex analysis.

---

## pivot_table()

### Purpose

Creates an Excel-like summary table by aggregating data across rows and columns.

### Syntax

```python
df.pivot_table(
    values='Salary',
    index='Department',
    columns='Gender',
    aggfunc='mean'
)
```

### Example

| Department | Gender | Salary |
|------------|--------|--------|
| IT | Male | 50000 |
| IT | Female | 60000 |
| HR | Male | 40000 |
| HR | Female | 45000 |

Output:

| Department | Female | Male |
|------------|---------|------|
| HR | 45000 | 40000 |
| IT | 60000 | 50000 |

---

## Difference

| groupby() | pivot_table() |
|------------|--------------|
| Groups data | Creates summary table |
| Similar to SQL GROUP BY | Similar to Excel Pivot Table |
| More flexible | Easier for reporting |
| Returns grouped object | Returns summarized DataFrame |
| Doesn't automatically create columns | Can create row and column summaries |

---

## When to Use groupby()

- Data aggregation
- Data transformation
- Custom functions
- Multiple aggregations
- Complex analysis

Example:

```python
df.groupby('Department').agg({
    'Salary':'mean',
    'Age':'max'
})
```

---

## When to Use pivot_table()

- Dashboard reporting
- Cross-tab reports
- Excel-style summaries
- Comparing multiple categories

---

## Common Interview Questions

### Which one is more flexible?

**Answer:** `groupby()` because it supports complex transformations and custom operations.

---

### Which one is easier for business reporting?

**Answer:** `pivot_table()` because it creates Excel-like reports.

---

### Can pivot_table() perform aggregation?

Yes.

Common aggregation functions:

- mean
- sum
- count
- min
- max
- median

---

## Interview Tip

Remember:

**groupby() → Analysis**

**pivot_table() → Reporting**

# Interview Question 10: Difference Between Merge, Join, and Concat

## Interview Question

**Explain the difference between `merge()`, `join()`, and `concat()` in Pandas.**

## Answer

All three functions combine DataFrames, but they are used in different situations.

---

# 1. merge()

### Purpose

Combines DataFrames using one or more common columns (keys).

Similar to **SQL JOIN**.

### Syntax

```python
pd.merge(df1, df2, on='Employee_ID')
```

### Types of Merge

- Inner
- Left
- Right
- Outer

### Example

Employee Table

| Employee_ID | Name |
|--------------|------|
| 1 | Aryan |
| 2 | Rahul |

Salary Table

| Employee_ID | Salary |
|--------------|--------|
| 1 | 50000 |
| 2 | 60000 |

```python
pd.merge(df1, df2, on='Employee_ID')
```

Output

| Employee_ID | Name | Salary |
|--------------|------|--------|
| 1 | Aryan | 50000 |
| 2 | Rahul | 60000 |

---

# 2. join()

### Purpose

Joins DataFrames using their **index** by default.

### Syntax

```python
df1.join(df2)
```

Can also join on a column using:

```python
df1.join(df2.set_index('Employee_ID'), on='Employee_ID')
```

### Features

- Faster for index-based joins.
- Simpler syntax than merge().
- Mostly used when indexes already match.

---

# 3. concat()

### Purpose

Stacks DataFrames together.

### Vertical Concatenation (axis=0)

```python
pd.concat([df1, df2])
```

Adds rows.

---

### Horizontal Concatenation (axis=1)

```python
pd.concat([df1, df2], axis=1)
```

Adds columns.

---

## Example

DataFrame 1

| Name |
|------|
| Aryan |
| Rahul |

DataFrame 2

| Name |
|------|
| Amit |
| Priya |

```python
pd.concat([df1, df2])
```

Output

| Name |
|------|
| Aryan |
| Rahul |
| Amit |
| Priya |

---

## Comparison Table

| merge() | join() | concat() |
|----------|---------|-----------|
| Combines using common columns | Combines using index | Stacks DataFrames |
| Similar to SQL JOIN | Index-based merge | Row/Column stacking |
| Most commonly used | Faster for index joins | No matching key required |
| Supports Inner, Left, Right, Outer | Mostly Left Join by default | Axis = 0 or 1 |

---

## When to Use Each?

### merge()

Use when:

- Common key exists
- SQL-style joins
- Relational datasets

Example:

Customers + Orders

---

### join()

Use when:

- Indexes match
- Index-based combination
- Faster joins

---

### concat()

Use when:

- Combining monthly datasets
- Combining multiple CSV files
- Appending rows
- Appending columns

---

## Common Interview Questions

### Which is similar to SQL JOIN?

**Answer:** `merge()`

---

### Which uses index by default?

**Answer:** `join()`

---

### Which one does not require a common column?

**Answer:** `concat()`

---

### Can concat() perform SQL joins?

**Answer:** No.

It simply stacks DataFrames without matching keys.

---

## Interview Tip

Remember this shortcut:

- **merge() → Match by Columns (SQL JOIN)**
- **join() → Match by Index**
- **concat() → Stack DataFrames**
