### 1️⃣ Introduction to NumPy

 If we go in depth is why NUMPY is faster than PYTHON :-
 1. Then the main point is NUMPY runs on C language internally all the operations in NUMPY is done in C which is compiled language.
 2. And second main point is in PYTHON elements in the LIST are not store in continuity which means for ex; ls = [1,2,3,4]

So, here in 'ls' the element '1' can be stored at let say memory address of 3000 but it is not necessary that '2' will be stored
at the 3001 address it can be at the any other address . Hence it becomes difficult for python to locate them in the Static memory

Whereas in NUMPY which runs on the C lang the elements of the array are stored in continuity.
Ex: a = [1 2 3 4] here if element '1' is stored at address = 3000 THEN element '2' will be at address = 3001
HEnce it becomes easier for the NUMPY to access the elements

## DATA TRANSFORMATIONS (USING WHERE AND SELECT)

## 1️⃣ np.where() — BASIC IDEA

```python
arr6 = np.array([10, 20, 30, 40])

result = np.where(arr6 > 25, "High", "Low")

print(result)
```

```python
arr7 = np.array([1,2,3,4,5])

result7 = np.where(arr7 % 2 == 0, 0, arr7) # here we are replcing even number with 0 and keeping odd numbers as it is

print(result7)
```

```python
# MODIFY EXISTING ARRAY
arr8 = np.array([10,20,30,40])

arr8 = np.where(arr8 > 20, arr8 * 2, arr8)

print(arr8)
```

```python
# MULTIPLE CONDITIONS USING WHERE (NESTED) - NOT RECOMMENDED
arr9 = np.array([10,20,30,40,50])

result9 = np.where(
            arr9 < 20, "Low",
            np.where(arr9 < 40, "Medium", "High")
          )

print(result9)
```

```python
arr10 = np.array([10,20,30,40])

np.where(arr10 > 25)
# it will return an array of indexes which will match this criteria
```

```python
salary = np.array([20000,35000,50000,15000])

grade = np.where(salary > 30000, "Good", "Bad")

print(grade)
```

# 🔷 np.select() — MULTIPLE CONDITIONS

```python
marks = np.array([35, 55, 75, 90])

conditions = [
    marks < 40,
    marks < 70,
    marks >= 70
]

choices = ["Fail", "Pass", "Distinction"]

result11 = np.select(conditions, choices, default="Unknown")

print(result11)
```

```python
age = np.array([12, 18, 25, 60])

conditions = [
    age < 18,
    age < 60,
    age >= 60
]

choices = [
    "Child",
    "Adult",
    "Senior"
]

result12 = np.select(conditions, choices,default="Unknown")

print(result12)
```

```python
purchase = np.array([100,500,2000,7000])

conditions = [
    purchase < 500,
    purchase < 3000,
    purchase >= 3000
]

choices = [
    "Low",
    "Medium",
    "High"
]

print(np.select(conditions, choices,default="Unknown"))
```

```python
marks = np.array([45,65,85])

conditions = [
    marks < 50,
    marks < 80,
    marks >= 80
]

choices = ["C","B","A"]

print(np.select(conditions, choices,default=""))
```
