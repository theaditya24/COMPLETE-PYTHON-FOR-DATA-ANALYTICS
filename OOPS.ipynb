# Object-Oriented Programming (OOP) in Python

**Why OOP?**
- Organizes code better
- Promotes code reusability
- Makes code easier to maintain and scale
- Models real-world entities

```python

# ═══════════════════════════════════════════════════════════════
# PYTHON OOP — CLASSES & OBJECTS  |  A complete reference cell
# ═══════════════════════════════════════════════════════════════

# ── 1. WHAT IS A CLASS? ──────────────────────────────────────
#   A class is a blueprint/template for creating objects.
#
#   Real Life          Python
#   ─────────────────────────
#   Car blueprint   →  Class
#   Actual car      →  Object
# ─────────────────────────────────────────────────────────────

class Dog:
    """This is a simple class.  Access via: Dog.__doc__"""

    # ── 2. CLASS ATTRIBUTE ───────────────────────────────────────
    #   Belongs to the CLASS itself, not individual objects.
    #   Shared by ALL dog objects.  Access: Dog.species  or  dog1.species
    species = "Canis familiaris"

    # ── 3. CONSTRUCTOR / __init__ ────────────────────────────────
    #   Runs AUTOMATICALLY when an object is created.
    #   You never call it manually.
    #   self = the current object being created.It points to the location of the object.
    def __init__(self, name, age):

        # ── 4. INSTANCE ATTRIBUTES ───────────────────────────────
        #   Object-specific variables — each object gets its OWN copy.
        #   self.name stores the value INSIDE the object permanently.
        #   Without self, the variable would vanish after __init__ ends.
        self.name = name
        self.age  = age

    # ── 5. INSTANCE METHODS ──────────────────────────────────────
    #   Functions defined inside a class; operate on objects.
    def bark(self):
        # f-string: inserts variables into strings
        print(f"{self.name} says Woof Woof!")

    def get_info(self):
        # return sends a value BACK so it can be stored or reused
        return f"{self.name} is {self.age} years old."


# ── 6. CREATING OBJECTS ──────────────────────────────────────
#   dog1 = Dog("Buddy", 5)  does 4 things:
#     1. Creates an empty object in memory
#     2. Calls  __init__(dog1, "Buddy", 5)  automatically
#     3. Stores  dog1.name = "Buddy"  and  dog1.age = 5
#     4. Returns the object and assigns it to dog1

dog1 = Dog("Buddy", 5)
dog2 = Dog("Max",   3)


# ── 7. ACCESSING ATTRIBUTES ──────────────────────────────────
print(dog1.name)          # → Buddy
print(dog2.name)          # → Max
print(dog2.age)           # → 3


# ── 8. CALLING METHODS ───────────────────────────────────────
#   dog1.bark()  →  Python converts this to:  Dog.bark(dog1)
#   so inside bark(),  self  becomes  dog1

dog1.bark()               # → Buddy says Woof Woof!
dog2.bark()               # → Max says Woof Woof!


# ── 9. return vs print ───────────────────────────────────────
#   print()  →  displays output, cannot be reused
#   return   →  gives value back, can be stored and reused

info = dog1.get_info()
print(info)               # → Buddy is 5 years old.
print(dog2.get_info())    # → Max is 3 years old.


# ── 10. CLASS ATTRIBUTE ACCESS ───────────────────────────────
print(Dog.species)        # → Canis familiaris  (from the class)
print(dog1.species)       # → Canis familiaris  (still the class's copy)


# ── VISUAL MEMORY REPRESENTATION ─────────────────────────────
#
#   Dog Class
#   ├── species = "Canis familiaris"   ← shared by all objects
#   ├── __init__()
#   ├── bark()
#   └── get_info()
#
#   dog1 object            dog2 object
#   ├── name = "Buddy"     ├── name = "Max"
#   └── age  = 5           └── age  = 3
#
# ── OOP GLOSSARY ──────────────────────────────────────────────
#   Class            →  Blueprint / template
#   Object/Instance  →  A real thing created from the class
#   Attribute        →  A variable inside a class or object
#   Method           →  A function defined inside a class
#   Constructor      →  __init__()  runs on object creation
#   self             →  Refers to the current object
#   Class attribute  →  Shared across ALL objects  (e.g. species)
#   Instance attr.   →  Unique to each object      (e.g. name, age)
# ═════════════════════════════════════════════════════════════
```

## Instance vs Class vs Static Methods

```python
# ═══════════════════════════════════════════════════════════════════════
# PYTHON OOP — INSTANCE, CLASS & STATIC METHODS  |  Complete reference
# ═══════════════════════════════════════════════════════════════════════

# ── 1. THE THREE METHOD TYPES — QUICK OVERVIEW ──────────────────────
#
#   Method Type     Decorator       First param    Works with
#   ──────────────────────────────────────────────────────────
#   Instance method  (none)          self           the object
#   Class method     @classmethod    cls            the class
#   Static method    @staticmethod   (none)         neither
# ─────────────────────────────────────────────────────────────────────

class Employee:
    """Blueprint for Employee objects."""

    # ── 2. CLASS VARIABLE ────────────────────────────────────────────────
    #   Belongs to the CLASS, shared by ALL Employee objects.
    #   Access: Employee.company  or  emp.company
    company = "TechCorp"

    # ── 3. CONSTRUCTOR ───────────────────────────────────────────────────
    #   Runs automatically when an object is created.
    #   self = the new object being built.
    def __init__(self, name, salary):
        self.name   = name    # instance attribute — unique per object
        self.salary = salary  # instance attribute — unique per object

    # ── 4. INSTANCE METHOD ───────────────────────────────────────────────
    #   • Takes  self  as the first parameter (the calling object)
    #   • Can read/write BOTH instance attributes AND class variables
    #   • Called on an object:  emp.show_details()
    #   • Python converts it to:  Employee.show_details(emp)
    def show_details(self):
        print(f"Name: {self.name}, Salary: {self.salary}")

    # ── 5. CLASS METHOD ──────────────────────────────────────────────────
    #   • Decorated with  @classmethod
    #   • Takes  cls  as the first parameter (the class itself, not object)
    #   • Can read/write CLASS variables only (no self, no instance data)
    #   • Called on the class:   Employee.change_company("xAI")
    #   • Can also be called on an object:  emp.change_company("xAI")
    #     → but it still receives the CLASS, not the object
    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company           # updates the class variable for ALL objects
        print(f"Company changed to {new_company}")

    # ── 6. STATIC METHOD ─────────────────────────────────────────────────
    #   • Decorated with  @staticmethod
    #   • Takes NO  self  or  cls  — it's a plain utility function
    #   • Cannot access instance attributes OR class variables
    #   • Belongs inside the class only because it's logically related
    #   • Called on the class:   Employee.is_valid_salary(50000)
    #   • Can also be called on an object:  emp.is_valid_salary(50000)
    @staticmethod
    def is_valid_salary(salary):
        return salary > 0    # returns True if salary is positive, else False


# ── 7. CREATING AN OBJECT ────────────────────────────────────────────
#   Calls __init__(emp, "Aditya", 75000) automatically
emp = Employee("Aditya", 75000)


# ── 8. CALLING AN INSTANCE METHOD ────────────────────────────────────
#   emp.show_details()  →  self becomes emp  →  reads emp.name, emp.salary
emp.show_details()                        # → Name: Aditya, Salary: 75000


# ── 9. CALLING A CLASS METHOD ────────────────────────────────────────
#   Preferred call: on the CLASS itself (makes intent clear)
#   cls.company = "xAI" updates the class variable → affects ALL objects
Employee.change_company("xAI")           # → Company changed to xAI
print(Employee.company)                  # → xAI
print(emp.company)                        # → xAI  (all objects see the change)


# ── 10. CALLING A STATIC METHOD ──────────────────────────────────────
#   No self, no cls — just a pure function living inside the class
print(Employee.is_valid_salary(50000))   # → True
print(Employee.is_valid_salary(-100))    # → False
print(emp.is_valid_salary(50000))        # → True  (works on object too)


# ── COMPARISON TABLE ─────────────────────────────────────────────────
#
#   Feature            Instance method   Class method      Static method
#   ─────────────────────────────────────────────────────────────────────
#   Decorator          (none)            @classmethod      @staticmethod
#   First parameter    self (object)     cls (class)       (none)
#   Access instance?   YES               NO                NO
#   Access class var?  YES               YES               NO
#   Typical use        object behaviour  factory/class op  utility/helper
#
# ── VISUAL MEMORY ────────────────────────────────────────────────────
#
#   Employee Class
#   ├── company = "xAI"          ← class variable (shared, mutable)
#   ├── __init__()               ← constructor
#   ├── show_details()           ← instance method  (needs self)
#   ├── change_company()         ← class method     (needs cls)
#   └── is_valid_salary()        ← static method    (needs nothing)
#
#   emp object
#   ├── name   = "Aditya"        ← instance attribute
#   └── salary = 75000           ← instance attribute
#
# ═════════════════════════════════════════════════════════════════════
```
