### Vertical Concatination

````python
summer = bios[bios['type'] == 'Summer']

winter = bios[bios['type'] == 'Winter']
````

````python
new_df = pd.concat([summer, winter])
````

````python
new_df.head()
````

````python
new_df.shape
````

````python
bios.shape
````

## Vertical Concatenation (`axis=0`)

Both `new_df` and `bios` have the **same number of rows** because `pd.concat()` simply stacks one DataFrame below another.

Since we split the original `bios` DataFrame into `summer` and `winter`, concatenating them vertically recreates the original dataset.

```python
new_df = pd.concat([summer, winter])
```

### Why is the number of rows unchanged?

- `summer` contains all Summer Olympics records.
- `winter` contains all Winter Olympics records.
- `pd.concat(..., axis=0)` stacks the rows of `winter` below the rows of `summer`.
- No rows are removed or duplicated during concatenation.

Therefore,

```python
new_df.shape == bios.shape
```

returns `True`.

---

## Problem with the Index

Although the rows are combined correctly, the **index is not reset**.

For example,

**summer**

| Index | Type |
|------:|------|
| 0 | Summer |
| 1 | Summer |
| 2 | Summer |

**winter**

| Index | Type |
|------:|------|
| 0 | Winter |
| 1 | Winter |
| 2 | Winter |

After concatenation,

| Index | Type |
|------:|------|
| 0 | Summer |
| 1 | Summer |
| 2 | Summer |
| 0 | Winter |
| 1 | Winter |
| 2 | Winter |

Notice that the index values are duplicated because Pandas **preserves the original indices** of both DataFrames.

This can lead to confusion when selecting rows using `.loc[]` or while performing further analysis.

---

## Solution: Reset the Index

Use `ignore_index=True` while concatenating.

```python
new_df = pd.concat([summer, winter], ignore_index=True)
```

Now the index becomes

| Index | Type |
|------:|------|
| 0 | Summer |
| 1 | Summer |
| 2 | Summer |
| 3 | Winter |
| 4 | Winter |
| 5 | Winter |

The rows remain the same, but the index is recreated sequentially from `0` to `n-1`.

---

### Key Takeaways

- `axis=0` → Adds **rows** (Vertical Concatenation)
- Number of columns remains unchanged.
- Total rows = Sum of rows in all DataFrames.
- Original indices are preserved by default.
- Use `ignore_index=True` to create a fresh sequential index.

## Horizontal Concat

````python
left = bios[['athlete_id', 'noc']].head()

right = bios[['year', 'medal']].head()
````

````python
combined = pd.concat(
    [left, right],
    axis=1
)

combined
````

### Rule
    Use Merge when there is a common column (key).

    Use Concat when you simply want to stack rows or place DataFrames side by side.
