```python
import numpy as np
import pandas as pd
```

```python
coffee = pd.read_csv("coffee.csv")
coffee.head()
```

```python
coffee.tail()
```

```python
coffee.sample()
```

```python
coffee.loc[0] # Syntax: df.loc[#rows, #columns
# Here we have not given any column value so it will take all the columns for 0th row
```

```python
coffee.loc[[0,3]] # 0th and 3rd index row
```

```python
coffee.loc[0:3] # from 0th to 3rd index row
```

```python
coffee.loc[0:3, "Day"]
```

```python
coffee.loc[0:3, ["Day","Units Sold"]] # here you can not use numbers to access columns 
# because .loc[] only gets that rows and cols which are equal to this value
```

```python
coffee.iloc[0:3,0:2] # Here You can use numbers to access both rows and cols because it gets the data by the index positions
# Also here the outer value is exclusive but in the .loc[] we saw it was inclusive.
```

```python
coffee.at[0,"Day"]
```
