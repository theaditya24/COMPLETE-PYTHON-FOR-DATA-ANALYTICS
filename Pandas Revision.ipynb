# Filtering Data

```python
bios = pd.read_csv("results.csv")
```

```python
bios.head()
```

```python
bios.info()
```

```python
# Find those athlete who have scored first place
bios.loc[bios['place'] == 1]
```

```python
# Find the athlete_id and medal of those athlete who have scored first place
bios.loc[bios['place'] == 1, ['athlete_id','medal']]
```

```python
# Find the athlete_id and medal of those athlete who have scored first place
bios[bios['place'] == 1][['athlete_id','medal']]
```

```python
# Find the athlete_id and medal of those athlete who have scored first place and also type of olymipics is summer
bios[ (bios['place'] == 1) & (bios['type'] == 'Summer') ][['athlete_id','medal']]
```

```python
bios[bios['name'].str.contains('sam')] # by this code you can find any name which contains 'sam' in it. 
```

```python
# Find those athletes whose country is wither 'USA' , 'FRA', 'UK' and their place is 1st and olympics type is summer
bios[ (bios['noc'].isin(["USA","FRA","UK"])) & (bios['place'] == 1) & (bios['type'] == 'Summer') ][['athlete_id','medal']]
```

```python
# Same above question can be done by using query()
bios.query('noc in ["USA", "FRA", "UK"] and place == 1 and type == "Summer"')[['athlete_id', 'medal']]
```

# ADDING / REMOVING COLUMNS

```python
coffee_new = coffee.copy() # if we do not use this copy() then both the coffee and coffee_new will point to same memory location
# where the data is stored and if we change in one variable it will lead to change in the another.
```

```python
coffee_new.head()
```

```python
coffee_new['price'] = 6.99
```

```python
coffee_new.head()
```

```python
coffee_new['Coffee Type'].value_counts()
```

```python
coffee_new['new_price'] = np.where(coffee_new['Coffee Type'] == 'Espresso', 4.99,6.99)
```

```python
coffee_new.head()
```

```python
coffee_new.drop(columns=['price'],inplace=True)
```

```python
coffee_new.head()
```

```python
coffee.head()
```

```python
coffee_new['revenue'] = coffee_new['Units Sold'] * coffee_new['new_price']
```

```python
coffee_new.head()
```

```python
coffee_new = coffee_new.rename(columns={'new_price':'Coffee Price', 'revenue':'Coffee Revenue'})
```
