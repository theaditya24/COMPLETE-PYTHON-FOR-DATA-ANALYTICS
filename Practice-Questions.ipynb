# 1. Text Analyzer Tool (BEST for learning)

### ✔ Count words
### ✔ Count characters
### ✔ Count sentences
### ✔ Most frequent word
### ✔ Remove punctuation

### Features:

### User enters a paragraph

### Your code analyzes it using loops + string functions + lists

```python
para = input("Enter any paragraph / word : ").strip()
# print(para)
# 1. Remove punctuation
punctuations = ".,!?;:-_()[]{}\"'"
clean_text = ""
for char in para:
    if char not in punctuations:
        clean_text += char
# print(clean_text)

# 2. Count Words
para_length = len(clean_text.split(" "))
print(f"Total number of words in the Para you entered are : {para_length}")

# 3. Count characters
char_count = 0
for char in clean_text:
    if char != " ":
        char_count += 1
print(f"Total number of Characters in the Para you entered are : {char_count}")

# 4. Count Sentences
sentence_count = 0
for char in para:
    if char in ".!?":
        sentence_count += 1
print(f"Total number of Sentences in the Para you entered are : {sentence_count}")

# 5. Most frequent word
freq = {}
for w in clean_text:
    w = w.lower()
    if w in freq:
        freq[w] += 1
    else:
        freq[w] = 1

# find word with highest count
most_frequent = max(freq, key=freq.get)
print("Most Frequent Word:", most_frequent)
```

```python
print("========================")
print("\tTO-DO LIST")
print("========================\n")

print('''1. Add Task
2. View Tasks
3. Remove Task
4. Exit\n''')

task = []

while True:
    task_number = int(input("Choose an option from the above : "))
    
    if task_number == 1:
        new_task = input("Enter a New Task you want to Add in To-Do list : ")
        task.append(new_task)
        print("\n Task Added Successfully!!")
        
    elif task_number == 2:
        if len(task) == 0:
            print("No tasks found!\n")
        else:
            print("\nYour Tasks are:")
            for i in range(len(task)):
                print(f"{i+1}. {task[i]}")
            print()
            
    elif task_number == 3:
        del_task = int(input("Enter the task number which you want to delete :"))
        if len(task) == 0:
            print("No tasks available to delete!\n")
        else:
            del_task = int(input("Enter the task number which you want to delete: "))
            if 1 <= del_task <= len(task):
                task.pop(del_task - 1)
                print("Task Deleted Successfully!\n")
            else:
                print("Invalid task number!\n")

    # 4. Exit
    elif task_number == 4:
        print("Exiting program... Thank you!")
        break

    else:
        print("Invalid option! Please enter 1–4.\n")
```

```python
def to_upper(word):
    result = ""
    for ch in word:
        if 'a' <= ch <= 'z':  
    # we have done this because if user enters any number or spec char or alph other than lower case then it will print the same
            result += chr(ord(ch) - 32)
        else:
            result += ch
    return result

print(to_upper("hello"))
```

### 2. Find the index of first non-repeating character in a word

```python
def first_non_repeat(word):
    freq = {}
    for char in word:
        if char in freq:
            freq[char] += 1
        else:
            freq[char] = 1
    for i in range(len(word)):
        if freq[word[i]] == 1:
            return i
    return -1
first_non_repeat("lovely")
```

### Find the longest word in the string/sentence

```python
# Input: I love programming
# Output: programming
def longest_Word(string):
    lst = string.split()
    freq = {}
    for word in lst:
        if word.lower() in freq:
            continue
        else:
            freq[word.lower()] = len(word)
            
    max_len = max(freq.values())
    for key, value in freq.items():
        if value == max_len:
            print(f'{key} -> {value}')
    
longest_Word("I love I programming")
```

```python
# simpler Approach
def longest_word(string):
    words = string.split()
    longest = ""

    for word in words:
        if len(word) > len(longest):
            longest = word

    return longest

print(longest_word("I love I programming"))
```

```python
# If there are multiple longest words and you want all of them:
def longest_words(sentence):
    words = sentence.split()
    max_len = max(map(len, words))
    return [word for word in words if len(word) == max_len]

print(longest_words("I love coding in Python"))
```

## 📝 String Compression

Write a Python function that compresses a string by replacing consecutive repeated characters with the character followed by its count.

If the compressed string is not shorter than the original string, return the original string.

### Example 1

**Input:** `aaabbccc`
**Output:** `a3b2c3`

### Example 2

**Input:** `aabcccccaaa`
**Output:** `a2b1c5a3`

### Example 3

**Input:** `abc`
**Output:** `abc`

**Explanation:** The compressed string would be `a1b1c1`, which is longer than the original string, so return the original string.


```python
# Basic Approach
def compress_string(s):
    compressed = ""
    count = 1

    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            count += 1
        else:
            compressed += s[i - 1] + str(count)
            count = 1

    compressed += s[-1] + str(count)

    if len(compressed) < len(s):
        return compressed

    return s

compress_string("aaabbbccaaa")

# Time Complexity: O(n²)
# Space Complexity: O(n)

# Why O(n²)? Because strings are immutable, and every += creates a new string.
```

```python
# Better Approach
def compress_string(s):
    result = []
    count = 1

    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            count += 1
        else:
            result.append(s[i - 1] + str(count))
            count = 1

    result.append(s[-1] + str(count))

    compressed = "".join(result)

    return compressed if len(compressed) < len(s) else s

compress_string("aaabbcc")
```

### Find all permutations of a string
#### Example : Input- str = "abc"
                output - ['abc',acb','bac','bca','cab','cba']

```python
def find_permutations(s):
    if len(s) == 1:
        return [s]

    permutation = []
    for i in range(len(s)):
        current_char = s[i]
        remaining_char = s[:i] + s[i+1:]

        remaining_permutations = find_permutations(remaining_char)
        for p in remaining_permutations:
            permutation.append(current_char + p)
    return permutation

find_permutations("abc")
```

    Avoid apply() when a vectorized Pandas operation exists because vectorized operations are faster and more efficient. 
    Use apply() only when custom row-wise or element-wise logic is needed.
