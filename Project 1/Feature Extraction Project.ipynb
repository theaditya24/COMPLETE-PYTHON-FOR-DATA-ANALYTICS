## Extract the Total Episodes from the title and save it to the New Column

```python
df.loc[2]['Title']
```

```python
# we want to make a new column of episode count and if we see into the title carefully episodes count is there in ()
def extract_count(txt):
    check = False
    data = ""
    for i in txt:
        if i == ')':
            break
        if i == '(':
            check = True
        if check == True:
            data += i
    return data
```

```python
df["Title"].apply(extract_count)
```

```python
ext_eps = df["Title"].apply(extract_count).str.replace('(', '', regex=False)
ext_eps
```

```python
final_ext_eps = ext_eps.str.replace(" eps","",regex=False)
final_ext_eps
```

```python
# → adds column temporarily in the current notebook/session.
df['Total episodes'] = final_ext_eps
```

```python
df
```

```python
# → saves the changes permanently into the CSV file.
df.to_csv('anime.csv', index=False)
```

## Make a new column for Time Stamp

```python
df
```

```python
df['Title'][2]
```

```python
def extract_timeStamp(txt):
    check = True
    data = ""
    for i in range(len(txt)):
        if txt[i] == ")":
            for j in range(i+1,i+20):
                data += txt[j]
            return data
```

```python
df['Title'].apply(extract_timeStamp)
```

```python
df['Title'].apply(extract_timeStamp).str.strip()
```

```python
df['Time Stamp'] = df['Title'].apply(extract_timeStamp).str.strip()
```

```python
df
```

```python
# to make changes permanent
df.to_csv('anime.csv', index=False)
```

```python
def total_months(time_stamp):
    
    months = {
        'Jan': 1, 'Feb': 2, 'Mar': 3, 'Apr': 4,
        'May': 5, 'Jun': 6, 'Jul': 7, 'Aug': 8,
        'Sep': 9, 'Oct': 10, 'Nov': 11, 'Dec': 12
    }
    
    parts = time_stamp.split()
    
    start_month = months[parts[0]]
    start_year = int(parts[1])
    
    end_month = months[parts[3]]
    end_year = int(parts[4])
    
    total = (end_year - start_year) * 12 + (end_month - start_month) + 1
    
    return total


df['Total_Months'] = df['Time Stamp'].apply(total_months)
```
