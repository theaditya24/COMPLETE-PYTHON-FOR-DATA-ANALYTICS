# Regex — Fundamentals

## 1. Definition

Regex (Regular Expression) is a pattern-based method used to search, find, extract, validate, replace, and clean text.

Regex is especially useful in Data Analytics for:

- Data cleaning
- Data preprocessing
- Text extraction
- Data validation
- String manipulation
- Feature engineering
- Exploratory Data Analysis
- Working with messy datasets

---

## 2. Why It Is Used

Real-world datasets often contain inconsistent text such as:

- Different phone number formats
- Different salary formats
- Invalid or inconsistent emails
- Unwanted special characters
- Extra spaces
- Mixed text and numbers
- Inconsistent categories

Regex can be used to identify and manipulate these patterns.

Example:

```text
₹50,000
INR 50000
Rs. 50,000
$60,000

### re.search() searches through the entire string and returns the first match it finds.

````python
text1 = "Aryan works as a Data Analyst"

result1 = re.search("Data", text1)

print(result1)
````

````python
### Getting the position - It returns the starting and ending positions of the match.
print(result1.span())
````

````python
text2 = "I am learning Python"

result2 = re.search(r"Java", text2)

print(result2)
````

````python
job_description = "We are looking for a Python Data Analyst"
python_result = re.search(r"Python", job_description)
sql_result = re.search(r"SQL", job_description)

if python_result:
    print("Python found")
else:
    print("Python not found")

if sql_result:
    print("SQL found")
else:
    print("SQL not found")
````

````python
job_description2 = "We are hiring a Python Data Analyst with SQL and Power BI experience."
python_result2 = re.search(r"Python", job_description2)
sql_result2 = re.search(r"SQL", job_description2)
power_bi = re.search(r"Power BI",job_description2)

if python_result2:
    print("Python found")
else:
    print("Python not found")

if sql_result2:
    print("SQL found")
else:
    print("SQL not found")

if power_bi:
    print("Power BI found")
else:
    print("Power BI not found")
````

#### Regex is case-sensitive by default.

````python
re.search("Python", "I am learning Python") # matches.
re.search("Python", "I am learning python") # does not match.
````

### Metacharacters
    Some characters have special meanings in regex.
    These are called metacharacters.
    Common regex metacharacters:
    
**`.  ^  $  *  +  ?  { }  [ ]  \  |  ( )`**

````python
sentence = """I love to analyze the data and
i used to analyse whatever happens around me
that's why i am doing data analysis analyzer reanalyse"""

result = re.findall(r"analy[sz]e", sentence)

# but here is the problem that if there is a word analyser or analyzing it will also get that
print(result)
````

````python
result_exact = re.findall(r"\banaly[sz]e\b", sentence)
result_exact
````
