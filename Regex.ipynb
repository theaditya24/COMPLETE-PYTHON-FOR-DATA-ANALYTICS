# Regex — Fundamentals

## 1. Definition

Regex (Regular Expression) is a pattern-based method used to search, find, extract, validate, replace, and clean text.

Regex is especially useful in Data Analytics for:

- Data cleaning
- Data preprocessing
- Text extraction
- Data validation
- String manipulation
- Feature engineering
- Exploratory Data Analysis
- Working with messy datasets

---

## 2. Why It Is Used

Real-world datasets often contain inconsistent text such as:

- Different phone number formats
- Different salary formats
- Invalid or inconsistent emails
- Unwanted special characters
- Extra spaces
- Mixed text and numbers
- Inconsistent categories

Regex can be used to identify and manipulate these patterns.

Example:

```text
₹50,000
INR 50000
Rs. 50,000
$60,000

### re.search() searches through the entire string and returns the first match it finds.

````python
text1 = "Aryan works as a Data Analyst"

result1 = re.search("Data", text1)

print(result1)
````

````python
### Getting the position - It returns the starting and ending positions of the match.
print(result1.span())
````

````python
text2 = "I am learning Python"

result2 = re.search(r"Java", text2)

print(result2)
````

````python
job_description = "We are looking for a Python Data Analyst"
python_result = re.search(r"Python", job_description)
sql_result = re.search(r"SQL", job_description)

if python_result:
    print("Python found")
else:
    print("Python not found")

if sql_result:
    print("SQL found")
else:
    print("SQL not found")
````

````python
job_description2 = "We are hiring a Python Data Analyst with SQL and Power BI experience."
python_result2 = re.search(r"Python", job_description2)
sql_result2 = re.search(r"SQL", job_description2)
power_bi = re.search(r"Power BI",job_description2)

if python_result2:
    print("Python found")
else:
    print("Python not found")

if sql_result2:
    print("SQL found")
else:
    print("SQL not found")

if power_bi:
    print("Power BI found")
else:
    print("Power BI not found")
````

#### Regex is case-sensitive by default.

````python
re.search("Python", "I am learning Python") # matches.
re.search("Python", "I am learning python") # does not match.
````

### Regex — Basic Regex Characters

### 1. Definition

Regex provides special characters called metacharacters that allow patterns to describe more than exact literal text.

Important metacharacters :

- `.`
- `^`
- `$`

---

### 2. Why They Are Used

These symbols help control:

- Which characters can match
- Where a pattern can occur
- Whether a pattern must appear at the beginning
- Whether a pattern must appear at the end

---

### 3. Syntax

```python

r"^pattern"
r"pattern$"

### 1. The **`.`** Metacharacter
#### Match any single character except a newline.

````python
text = "cat bat hat dog"

result = re.findall(r".at", text)

print(result)
````

````python
text = "cat bat hat dog"

result = re.findall(r"..t", text)

print(result)
````

### 2. The ^ Anchor
#### This is called the start-of-string anchor.

````python
string = "cat is an animal"
result = re.findall(r"^cat",string)
result
````

````python
password = input("Enter your Password : ")
Pass_Checker = r"^(?=.*[A-Z])(?=.*[a-z])(?=.*\d)(?=.*[@#%!\^&()\{\}\[\]])[A-Za-z\d@#%!\^&()\{\}\[\]]{8,15}$"

if re.fullmatch(Pass_Checker,password):
    print("Valid Password")
else:
    print("Not a valid password")
````

### Password Regex Breakdown

1. `^`  
   → Start of the password

2. `(?=.*[A-Z])`  
   → Must contain at least 1 uppercase letter

3. `(?=.*[a-z])`  
   → Must contain at least 1 lowercase letter

4. `(?=.*\d)`  
   → Must contain at least 1 digit

5. `(?=.*[@#%!\^&()\{\}\[\]])`  
   → Must contain at least 1 allowed symbol

6. `[A-Za-z\d@#%!\^&()\{\}\[\]]{8,15}`  
   → Only allowed characters  
   → Password length must be between 8 and 15 characters

7. `$`  
   → End of the password


### How to Read the Regex

`^`  
→ START

`(?=.*[A-Z])`  
→ Must contain uppercase

`(?=.*[a-z])`  
→ Must contain lowercase

`(?=.*\d)`  
→ Must contain digit

`(?=.*[@#%!\^&()\{\}\[\]])`  
→ Must contain special character

`[A-Za-z\d@#%!\^&()\{\}\[\]]{8,15}`  
→ Only allowed characters  
→ Length: 8–15

`$`  
→ END


### Two Use Cases of `^` in Regex

`^abc`  
→ String starts with `abc`

`[^abc]`  
→ Anything except `a`, `b`, or `c`
