# Sampling in Pandas

### Sampling means selecting some rows from a dataset.

```python
import pandas as pd

df = pd.DataFrame({
    'Name':['A','B','C','D','E','F'],
    'Marks':[85,90,78,92,88,95]
})

print(df)
```

### Type 1: Random Sampling by Number of Rows

```python
df.sample( n = 3 ) # Take 3 random rows
```

### Type 2: Fraction Sampling

```python
df.sample(frac = 0.5)  # Take 50% of rows
```

### Type 3: Sampling With Replacement

```python
df.sample(n=4, replace=True)   # Same row can be selected multiple times.
```

### Type 4: Sampling Without Replacement (Default)

```python
df.sample(n=3, replace=False)  # Each row appears once.
```

# Resampling in Pandas

### Resampling is used with date-time data to change frequency.

```python
import pandas as pd

data = {
    'Sales':[100,120,150,130,200,190,200,206]
}

dates = pd.date_range('2026-01-01', periods=8)

dt = pd.DataFrame(data, index=dates)

dt
```

### Downsampling - Converting high frequency data into lower frequency.

```python
# Daily → Monthly
dt.resample('ME').sum()
```

### Resampling Functions

#### 1. Sum

```python
dt.resample('ME').sum()
```

#### 2. Mean

```python
dt.resample('ME').mean()
```

#### 3. Max & Min

```python
# dt.resample('ME').max()
dt.resample('ME').min()
```

#### 4. Count

```python
dt.resample('ME').count()
```

### Upsampling - Lower frequency → Higher frequency

```python
df = pd.DataFrame(
    {'Sales':[100,200,300]},
    index=pd.to_datetime([
        '2025-01-01',
        '2025-02-01',
        '2025-03-01'
    ])
)
df
```

#### Create Daily Rows

```python
df.resample('D').asfreq()
```

#### Fill Missing Values After Upsampling
    Forward Fill

```python
df.resample('D').ffill()
```

#### Backward Fill

```python
df.resample('D').bfill()
```

#### Interpolation

```python
df.resample('D').interpolate()
# interpolate() = basically checks how much is the difference between one row and the another in the original df
# then it divides that difference by the number of days between them and then adds that number to the last value
```
