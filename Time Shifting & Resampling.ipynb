### Resampling

#### Resampling is a processs to convert the frequency

```python
import pandas as pd
import numpy as np

np.random.seed(42)

dates = pd.date_range(
    start='2026-01-01',
    periods=80,
    freq='D'
)

sales = np.random.randint(
    100,
    1000,
    size=80
)

df = pd.DataFrame({
    'Date': dates,
    'Sales': sales
})

df.set_index('Date', inplace=True)

df
```

```python
df.resample('M')
```

```python
list(df.resample('M'))
```

```python
df.resample('M').sum()   # day wise data ----> monthly sum    high frequency data ----> low frequency data  = downsampling
```

```python
df.resample('W').sum()
```

```python
df.resample('Y').sum()
```

```python
df
```
