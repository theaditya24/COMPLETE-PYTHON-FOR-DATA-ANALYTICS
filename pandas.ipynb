# Handling Missing Data In Pandas

```python
data_missing = {
    'A': [1,2, np.nan, 4, 5],
    'B': [1,2,3,4,5],
    'C': [1, 2, 3, np.nan,np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}

miss_data = pd.DataFrame(data_missing)
miss_data
```

### 1. Finding Missing Data = using 'isna()'

```python
miss_data.isna()
```

```python
miss_data.isna().sum() # Output will be the sum of how many NaN values each column contains.
```

```python
miss_data.isna().any() # Output: Here i want that if any the columns have atleast one NaN values
```

### 2. Removing Missing Values = It works on the basis of Rows

```python
miss_data.dropna() # Since all the rows have atleast one NaN values except row = 0 so all the data will be removed and only row = 0 will remain
```

```python
miss_data.dropna(thresh = 3)  # thresh = any_value it basically checks that if the rows have that usch amount of NON NULL values in it
```

### 3. Filling the missing Data

```python
miss_data.fillna(0)  # It will fill all the NaN values with '0' but the change is temporary to make it permanent we will use 'inplace=True'
```

```python
miss_data
```

```python
miss_data.fillna(value = 100)  # it will replace all the NaN with 100
```

```python
values = {'A':0,'B':100,'C':200,'D':150}
miss_data.fillna(value = values)  # since A and B columns does not have any NaN value so it will get ignored
```

```python
miss_data
```

```python
# Forward Fill (ffill) = 👉 Fills missing values using previous (above) 
miss_data.ffill()

# ✅ Output Explanation:
# Column A → NaN at index 2 becomes 2.0
# Column C → NaN becomes 3.0
# Column D → fills downward from 1.0
```

```python
# Backward Fill (bfill) = 👉 Fills missing values using next (below) value
miss_data.bfill()

# ✅ Output Explanation:
# Column A → NaN at index 2 becomes 4.0
# Column C → NaN becomes next valid (none below → stays NaN at end)
# Column D → all NaNs become 5.0
```

```python
miss_data.ffill().bfill()
```

# Merging , Joining And Concatination

```python
# DataFrame 1: Customers
df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Amit", "Riya", "John", "Sara"],
    "city": ["Delhi", "Mumbai", "Pune", "Delhi"]
})

# DataFrame 2: Orders
df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105],
    "customer_id": [1, 2, 2, 5, 6],
    "product": ["Laptop", "Mobile", "Tablet", "Camera", "Watch"]
})
```

```python
df_customers
```

```python
df_orders
```

### 1. MERGING (Most Important 🔥)

```python
# 🔹 Inner Join (Common records only)
pd.merge(df_customers, df_orders, on="customer_id", how="inner")
```

```python
# 🔹 Left Join
pd.merge(df_customers, df_orders, on="customer_id", how="left")

# 👉 All customers + matching orders
# 👉 Customers without orders → NaN
```

```python
# 🔹 Right Join
pd.merge(df_customers, df_orders, on="customer_id", how="right")

# 👉 All orders + matching customers
# 👉 Orders with no customer → NaN
```

```python
# 🔹 Outer Join (Full Join)
pd.merge(df_customers, df_orders, on="customer_id", how="outer")
```

### 2. JOIN (Index-based)

```python
# First set index:

df1 = df_customers.set_index("customer_id")
df2 = df_orders.set_index("customer_id")
```

```python
# 🔹 Left Join using .join()
df1.join(df2, how="left")

# 👉 Works similar to merge but uses index
```

### 3. CONCATENATION

```python
# 🔹 Row-wise (Stack rows)
pd.concat([df_customers, df_customers])

# 👉 Doubles the rows
```

```python
# 🔹 Column-wise (Side by side)
pd.concat([df_customers, df_orders], axis=1)

# 👉 Joins columns (may create NaN if sizes differ)
```

# GROUP BY AND AGGREGATION

```python
df_groupby = pd.DataFrame({
    "Department": ["IT", "HR", "IT", "HR", "Finance", "IT"],
    "Employee": ["A", "B", "C", "D", "E", "F"],
    "Salary": [60000, 50000, 70000, 45000, 80000, 65000],
    "Experience": [2, 3, 5, 2, 7, 4]
})

df_groupby
```

```python
# 🔹  Basic Grouping

# 👉 Group by single column
df_groupby.groupby("Department")
# will not show output until an aggregation function is used
```

```python
list( df_groupby.groupby("Department") )
```

### 🔹 Aggregation Functions

```python
# 👉 Sum
df_groupby.groupby("Department")["Salary"].sum().reset_index()
```

```python
# 👉 Mean
# df_groupby.groupby("Department")["Salary"].mean()

print()
# 👉 Count
# df_groupby.groupby("Department")["Employee"].count()

print()
# 👉 Multiple aggregations
df_groupby.groupby("Department")["Salary"].agg(["sum", "mean", "max", "min"])
# here it means that we have grouped the table using Department Column and then we have applied the aggregation functions on
# the salary column.
# other agg fun are :- median , count
```

### 🔹 Group by Multiple Columns

```python
df_groupby.groupby(["Department", "Experience"])["Salary"].mean()
```

### 🔹 Reset Index (Very Important)

```python
df_groupby.groupby("Department")["Salary"].mean().reset_index()
```

```python
# | Before reset_index() | After reset_index() |
# | -------------------- | ------------------- |
# | Department = index   | Department = column |
# | Harder to use        | Easy to use         |

# 🔹 Quick Memory Trick
# groupby() → column becomes index
# reset_index() → index becomes column again
```

### 🔹 Aggregating Multiple Columns

```python
df_groupby.groupby("Department").agg({
    "Salary": ["sum", "mean"],
    "Experience": "max"
})
```

### 🔹 Named Aggregation (Clean Output 🔥)

```python
df_groupby.groupby("Department").agg(
    avg_salary=("Salary", "mean"),
    total_salary=("Salary", "sum")
)
```

### 🔹 count() vs size()

```python
# df_groupby.groupby("Department").count()
df_groupby.groupby("Department").size()

# ✔️ Difference:

# count() → ignores NaN
# size() → counts all rows
```

### 🔹 value_counts() (Shortcut)

```python
df_groupby["Department"].value_counts()
```

### 🔹 3. Create Sample Dataset

```python
df_pvt= pd.DataFrame({
    "Date": ["2024-01", "2024-01", "2024-02", "2024-02", "2024-03"],
    "City": ["Delhi", "Mumbai", "Delhi", "Mumbai", "Delhi"],
    "Product": ["Laptop", "Laptop", "Mobile", "Mobile", "Laptop"],
    "Sales": [50000, 60000, 30000, 40000, 70000],
    "Quantity": [5, 6, 3, 4, 7]
})

df_pvt
```

### 🔹 4. Basic Pivot Table

```python
pd.pivot_table(df_pvt, values="Sales", index="City")
# 👉 Output: Average sales per city
# Here is the little catch that here we haven't give any aggregate function means we haven't specified any agg. fun()
# But the question is 🔹 Why is it giving average (mean)?

# 👉 Because pivot_table() has a default aggregation function
# ✅ Default: aggfunc = "mean"

# 🔹 So internally, your code behaves like:

# pd.pivot_table(df, values="Sales", index="City", aggfunc="mean")

# 👉 That’s why you get average sales per city

# So if you will not specify which aggfunc to use. It will use 'mean' as a by default aggfunc

# 🔹 Why Pandas chose mean as default?

# Because:
# Average is the most commonly used summary
# Same default behavior as many statistical tools

# 🔹 Common Mistake (Interview Trap ⚠️)

# Many students think:

# “Pivot table is just grouping”

# ❌ Wrong
# ✔️ It always aggregates, even if you don’t specify
```

### 🔹 5. Key Parameters (VERY IMPORTANT)

```python
pd.pivot_table(
    df_pvt,
    values="Sales",
    index="City",
    columns="Product",
    aggfunc="sum"
)

# 🔑 Meaning:
# values → column to aggregate
# index → rows (grouping)
# columns → columns (grouping)
# aggfunc → function (sum, mean, etc.)
```

### 🔹 6. Multiple Aggregations

```python
pd.pivot_table(
    df_pvt,
    values="Sales",
    index="City",
    aggfunc=["sum", "mean", "max"]
)
```
